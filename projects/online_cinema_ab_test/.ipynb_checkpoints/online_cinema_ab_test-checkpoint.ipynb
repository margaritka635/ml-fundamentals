# A/B-тестирование рекомендательной ленты онлайн-кинотеатра

**Задача:** понять, стоит ли выкатывать новую персональную рекомендательную ленту на всех пользователей.

В проекте есть полный цикл A/B-теста: постановка гипотез, генерация реалистичных данных, проверка рандомизации, AA-тест, основные статистические тесты, bootstrap, сегментный анализ, оценка мощности и финальное бизнес-решение.

## 1. Бизнес-контекст

Онлайн-кинотеатр тестирует новую главную страницу. В старой версии пользователи видят обычную подборку фильмов. В новой версии пользователи видят персональные рекомендации.

Ожидание бизнеса: новая лента должна чаще доводить пользователя до просмотра и увеличить вовлеченность.

Главная метрика: конверсия в просмотр.

Дополнительные метрики: время просмотра, возвращение на следующий день, выручка на пользователя.

Guardrail-метрика: доля жалоб. Она не должна вырасти.

## 2. План анализа

1. Сгенерировать реалистичные данные эксперимента.
2. Проверить качество данных и рандомизацию.
3. Провести AA-тест на контрольной группе.
4. Провести A/B-тест по главной метрике.
5. Проверить дополнительные метрики.
6. Использовать bootstrap для выручки и времени просмотра.
7. Использовать CUPED для снижения шума.
8. Применить поправку на множественные сравнения.
9. Посмотреть эффект по сегментам.
10. Оценить мощность и нужный размер выборки.
11. Принять бизнес-решение.

## 3. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, статистики и воспроизводимости результатов.

```
# pandas нужен для работы с таблицами.
import pandas as pd

# numpy нужен для генерации данных и расчетов.
import numpy as np

# matplotlib и seaborn нужны для графиков.
import matplotlib.pyplot as plt
import seaborn as sns

# scipy.stats нужен для статистических тестов.
from scipy import stats
from scipy.stats import chi2_contingency, ttest_ind

# Фиксирую seed, чтобы данные генерировались одинаково при повторном запуске.
np.random.seed(42)

# Настраиваю графики.
%matplotlib inline
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 4. Генерация данных

Данные синтетические, но сделаны похожими на продуктовый A/B-тест. Есть пользователи, группы control и treatment, устройства, страны, возрастные группы, метрики просмотра, возвращения, выручки и жалоб.

```
# Количество пользователей в эксперименте.
n = 40000

# Создаю user_id.
user_id = np.arange(1, n + 1)

# Рандомно делю пользователей на две группы.
group = np.random.choice(['control', 'treatment'], size=n, p=[0.5, 0.5])

# Генерирую пользовательские сегменты.
device = np.random.choice(['mobile', 'desktop', 'tablet'], size=n, p=[0.62, 0.30, 0.08])
country = np.random.choice(['RU', 'KZ', 'BY', 'AM'], size=n, p=[0.70, 0.14, 0.10, 0.06])
age_group = np.random.choice(['18-24', '25-34', '35-44', '45+'], size=n, p=[0.24, 0.38, 0.24, 0.14])
user_type = np.random.choice(['new', 'old'], size=n, p=[0.28, 0.72])

# Дополнительные признаки пользователя.
tenure_days = np.where(user_type == 'old', np.random.randint(60, 900, size=n), np.random.randint(1, 60, size=n))
subscription_plan = np.random.choice(['free', 'basic', 'premium'], size=n, p=[0.46, 0.38, 0.16])

# Pre-period метрика: сколько минут пользователь смотрел до эксперимента.
pre_watch_time = np.random.gamma(shape=2.0, scale=14.0, size=n)
pre_watch_time += np.where(user_type == 'old', 12, 0)
pre_watch_time += np.where(subscription_plan == 'premium', 18, 0)
pre_watch_time = np.clip(pre_watch_time, 0, None)

# Базовая вероятность просмотра зависит от типа пользователя и устройства.
base_watch = 0.38
base_watch += np.where(user_type == 'old', 0.08, 0.00)
base_watch += np.where(device == 'mobile', 0.03, 0.00)
base_watch += np.where(device == 'tablet', -0.02, 0.00)
base_watch += np.where(subscription_plan == 'premium', 0.05, 0.00)

# Новая рекомендательная лента немного повышает вероятность просмотра.
effect_watch = np.where(group == 'treatment', 0.025, 0.00)
watch_prob = np.clip(base_watch + effect_watch, 0.01, 0.95)
watched_video = np.random.binomial(1, watch_prob)

# Время просмотра есть в основном у тех, кто начал смотреть.
watch_time = np.where(watched_video == 1, np.random.gamma(shape=2.2, scale=18, size=n), 0)
watch_time += 0.18 * pre_watch_time * watched_video
watch_time += np.where((group == 'treatment') & (watched_video == 1), np.random.normal(3.0, 6.0, size=n), 0)
watch_time = np.clip(watch_time, 0, None)

# Возвращение на следующий день зависит от просмотра и немного от новой ленты.
ret_prob = 0.18 + 0.18 * watched_video + np.where(group == 'treatment', 0.012, 0.00)
returned_next_day = np.random.binomial(1, np.clip(ret_prob, 0.01, 0.95))

# Выручка скошенная: у большинства 0, у части пользователей есть платежи.
pay_prob = 0.055 + 0.035 * watched_video + np.where(subscription_plan == 'premium', 0.04, 0.00) + np.where(group == 'treatment', 0.006, 0.00)
paid = np.random.binomial(1, np.clip(pay_prob, 0.01, 0.50))
revenue = paid * np.random.lognormal(mean=3.2, sigma=0.7, size=n)
revenue += np.where((group == 'treatment') & (paid == 1), np.random.lognormal(mean=1.1, sigma=0.4, size=n), 0)

# Жалобы как guardrail. Специально закладываю маленький рост, чтобы его проверить.
complaint_prob = 0.012 + np.where(group == 'treatment', 0.002, 0.00)
complaint = np.random.binomial(1, complaint_prob)

# Дата попадания в эксперимент.
date = pd.to_datetime('2026-03-01') + pd.to_timedelta(np.random.randint(0, 14, size=n), unit='D')

# Собираю таблицу.
df = pd.DataFrame({
    'user_id': user_id,
    'date': date,
    'group': group,
    'device': device,
    'country': country,
    'age_group': age_group,
    'user_type': user_type,
    'tenure_days': tenure_days,
    'subscription_plan': subscription_plan,
    'pre_watch_time': pre_watch_time,
    'watched_video': watched_video,
    'watch_time': watch_time,
    'returned_next_day': returned_next_day,
    'revenue': revenue,
    'complaint': complaint
})

print(f'Размер данных: {df.shape[0]} строк, {df.shape[1]} колонок')
df.head()
```

**Короткий вывод:** данные похожи на реальный продуктовый эксперимент: есть группы, пользовательские сегменты и несколько метрик разного типа.

## 5. Проверка качества данных

Перед тестами нужно убедиться, что нет дублей пользователей, пропусков и пересечения пользователей между группами.

```
# Проверяю пропуски и дубли.
print('Пропусков всего:', df.isna().sum().sum())
print('Дублей user_id:', df['user_id'].duplicated().sum())

# Проверяю, попал ли пользователь более чем в одну группу.
user_groups = df.groupby('user_id')['group'].nunique()
print('Пользователей в двух группах:', (user_groups > 1).sum())
```

## 6. Проверка размера групп

Сначала смотрю, насколько ровно пользователи распределились между control и treatment. Сильный перекос может быть признаком проблемы с рандомизацией.

```
# Считаю размер групп.
group_size = df['group'].value_counts().rename_axis('group').reset_index(name='users')
group_size['share'] = group_size['users'] / len(df)
group_size
```

```
# Проверяю Sample Ratio Mismatch через хи-квадрат.
expected = np.array([len(df) / 2, len(df) / 2])
observed = df['group'].value_counts().sort_index().values
chi2 = ((observed - expected) ** 2 / expected).sum()
p_srm = 1 - stats.chi2.cdf(chi2, df=1)

print(f'chi2: {chi2:.4f}')
print(f'p-value SRM: {p_srm:.4f}')

if p_srm < 0.05:
    print('Есть подозрение на перекос размера групп.')
else:
    print('Сильного перекоса размера групп нет.')
```

## 7. Проверка баланса сегментов

Проверяю, похожи ли группы по устройствам, странам, возрасту и типу пользователя. Это не доказывает идеальную рандомизацию, но помогает найти грубые проблемы.

```
# Функция для проверки баланса категориального признака.
def segment_balance(col):
    tab = pd.crosstab(df[col], df['group'], normalize='columns')
    return (tab * 100).round(2)

# Баланс по устройствам.
segment_balance('device')
```

```
# Баланс по странам.
segment_balance('country')
```

```
# Баланс по возрастным группам.
segment_balance('age_group')
```

**Короткий вывод:** если доли сегментов в control и treatment близки, рандомизация выглядит нормально. Дальше можно переходить к тестам метрик.

## 8. Общий обзор метрик

Смотрю средние значения метрик по группам. Это еще не статистический вывод, а первичная аналитика.

```
# Сравниваю метрики по группам.
summary = df.groupby('group').agg(
    users=('user_id', 'count'),
    watch_rate=('watched_video', 'mean'),
    avg_watch_time=('watch_time', 'mean'),
    return_rate=('returned_next_day', 'mean'),
    arpu=('revenue', 'mean'),
    complaint_rate=('complaint', 'mean')
).round(4)

summary
```

Проверяю, нет ли странной динамики во времени. В реальном эксперименте важно смотреть не только общий итог, но и то, как метрики менялись по дням.

```
# Считаю дневные метрики по группам.
daily = df.groupby(['date', 'group']).agg(
    users=('user_id', 'count'),
    watch_rate=('watched_video', 'mean'),
    avg_watch_time=('watch_time', 'mean'),
    arpu=('revenue', 'mean'),
    complaint_rate=('complaint', 'mean')
).reset_index()

daily.head()
```

```
# Строю динамику главной метрики по дням.
sns.lineplot(data=daily, x='date', y='watch_rate', hue='group', marker='o')
plt.title('Динамика конверсии в просмотр по дням')
plt.xlabel('Дата')
plt.ylabel('Watch rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Короткий вывод:** дневная динамика помогает заметить технические сбои, перекосы запуска или эффект новизны. Если эффект есть только в один день, к нему нужно относиться осторожно.

## 10. AA-тест

Перед A/B-тестом полезно сделать AA-тест: взять только control, случайно разделить его на две части и проверить, что между ними нет значимой разницы. Если AA-тест часто показывает эффект, значит с данными или тестом что-то не так.

```
# Беру контрольную группу и случайно делю ее на две части.
control_df = df[df['group'] == 'control'].copy()
control_df['aa_group'] = np.random.choice(['A1', 'A2'], size=len(control_df), p=[0.5, 0.5])

# Считаю конверсию в просмотр внутри AA-групп.
aa_summary = control_df.groupby('aa_group')['watched_video'].mean().round(4)
aa_summary
```

```
# Хи-квадрат тест для AA по конверсии в просмотр.
aa_tab = pd.crosstab(control_df['aa_group'], control_df['watched_video'])
chi2, p_aa, dof, expected = chi2_contingency(aa_tab)

print(f'p-value AA-теста: {p_aa:.4f}')

if p_aa < 0.05:
    print('AA-тест нашел различие. Это тревожный сигнал.')
else:
    print('AA-тест не нашел различий. Можно идти дальше.')
```

## 11. Функции для A/B-тестов

Пишу небольшие функции, чтобы дальше не повторять один и тот же код для разных метрик.

```
# Функция для сравнения конверсий через z-test.
def proportion_z_test(success_a, n_a, success_b, n_b):
    p_a = success_a / n_a
    p_b = success_b / n_b
    p_pool = (success_a + success_b) / (n_a + n_b)
    se = np.sqrt(p_pool * (1 - p_pool) * (1 / n_a + 1 / n_b))
    z = (p_b - p_a) / se
    p_value = 2 * (1 - stats.norm.cdf(abs(z)))
    return p_a, p_b, p_b - p_a, z, p_value

# Функция для bootstrap-интервала разницы средних.
def bootstrap_diff(a, b, n_boot=3000, seed=42):
    rng = np.random.default_rng(seed)
    diffs = []
    a = np.array(a)
    b = np.array(b)
    for i in range(n_boot):
        a_sample = rng.choice(a, size=len(a), replace=True)
        b_sample = rng.choice(b, size=len(b), replace=True)
        diffs.append(b_sample.mean() - a_sample.mean())
    diffs = np.array(diffs)
    return diffs.mean(), np.percentile(diffs, 2.5), np.percentile(diffs, 97.5)
```

## 12. Главная метрика: конверсия в просмотр

Проверяю, выросла ли доля пользователей, которые начали смотреть видео. Это главная метрика эксперимента.

```
# Считаю данные для z-test.
a = df[df['group'] == 'control']['watched_video']
b = df[df['group'] == 'treatment']['watched_video']

p_a, p_b, diff, z, p_watch = proportion_z_test(a.sum(), len(a), b.sum(), len(b))
uplift_watch = diff / p_a

print(f'Control watch rate: {p_a:.4f}')
print(f'Treatment watch rate: {p_b:.4f}')
print(f'Absolute diff: {diff:.4f}')
print(f'Relative uplift: {uplift_watch:.2%}')
print(f'z-statistic: {z:.4f}')
print(f'p-value: {p_watch:.6f}')
```

**Короткий вывод:** если p-value меньше 0.05 и uplift положительный, новая лента статистически значимо повышает конверсию в просмотр.

## 13. Время просмотра

Время просмотра — непрерывная метрика. Сначала использую t-test, а потом bootstrap, потому что распределение времени обычно скошено.

```
# Разделяю время просмотра по группам.
a = df[df['group'] == 'control']['watch_time']
b = df[df['group'] == 'treatment']['watch_time']

# t-test для средних.
t_stat, p_time = ttest_ind(a, b, equal_var=False)

print(f'Control avg watch_time: {a.mean():.2f}')
print(f'Treatment avg watch_time: {b.mean():.2f}')
print(f'Diff: {b.mean() - a.mean():.2f}')
print(f'p-value t-test: {p_time:.6f}')
```

```
# Bootstrap для разницы среднего времени просмотра.
time_mean, time_left, time_right = bootstrap_diff(a, b)

print(f'Bootstrap diff: {time_mean:.2f}')
print(f'95% CI: [{time_left:.2f}, {time_right:.2f}]')
```

## 14. CUPED для времени просмотра

CUPED использует метрику до эксперимента, чтобы снизить шум. Если pre-period метрика связана с основной метрикой, тест становится чувствительнее.

```
# Функция CUPED-коррекции.
def cuped(y, covariate):
    theta = np.cov(y, covariate, ddof=1)[0, 1] / np.var(covariate, ddof=1)
    y_adj = y - theta * (covariate - covariate.mean())
    return y_adj, theta

# Считаю CUPED для watch_time.
y_metric = df['watch_time'].values
x_cov = df['pre_watch_time'].values
y_adj, theta = cuped(y_metric, x_cov)
df['watch_time_cuped'] = y_adj

print(f'theta: {theta:.4f}')
print('Дисперсия до CUPED:', round(np.var(y_metric), 2))
print('Дисперсия после CUPED:', round(np.var(y_adj), 2))
```

```
# Сравниваю группы по CUPED-скорректированной метрике.
a = df[df['group'] == 'control']['watch_time_cuped']
b = df[df['group'] == 'treatment']['watch_time_cuped']

t_stat, p_time_cuped = ttest_ind(a, b, equal_var=False)
cuped_mean, cuped_left, cuped_right = bootstrap_diff(a, b)

print(f'CUPED diff: {b.mean() - a.mean():.2f}')
print(f'p-value CUPED t-test: {p_time_cuped:.6f}')
print(f'CUPED bootstrap 95% CI: [{cuped_left:.2f}, {cuped_right:.2f}]')
```

**Короткий вывод:** CUPED полезен, когда есть хорошая метрика до эксперимента. Он не меняет смысл эффекта, но может сделать оценку стабильнее.

## 15. Возвращение на следующий день

Retention — бинарная метрика, поэтому снова использую z-test для долей.

```
# Сравниваю returned_next_day.
a = df[df['group'] == 'control']['returned_next_day']
b = df[df['group'] == 'treatment']['returned_next_day']

p_a, p_b, diff, z, p_return = proportion_z_test(a.sum(), len(a), b.sum(), len(b))

print(f'Control return rate: {p_a:.4f}')
print(f'Treatment return rate: {p_b:.4f}')
print(f'Absolute diff: {diff:.4f}')
print(f'Relative uplift: {diff / p_a:.2%}')
print(f'p-value: {p_return:.6f}')
```

## 16. Выручка на пользователя

Выручка обычно имеет много нулей и длинный хвост. Поэтому для нее особенно полезен bootstrap.

```
# Сравниваю revenue.
a = df[df['group'] == 'control']['revenue']
b = df[df['group'] == 'treatment']['revenue']

# t-test оставляю как ориентир.
t_stat, p_revenue = ttest_ind(a, b, equal_var=False)

# Bootstrap-интервал.
rev_mean, rev_left, rev_right = bootstrap_diff(a, b)

print(f'Control ARPU: {a.mean():.3f}')
print(f'Treatment ARPU: {b.mean():.3f}')
print(f'Diff: {b.mean() - a.mean():.3f}')
print(f'p-value t-test: {p_revenue:.6f}')
print(f'Bootstrap 95% CI: [{rev_left:.3f}, {rev_right:.3f}]')
```

## 17. Guardrail-метрика: жалобы

Жалобы не должны вырасти. Если новая лента повышает основные метрики, но сильно увеличивает жалобы, раскатывать ее опасно.

```
# Сравниваю complaint rate.
a = df[df['group'] == 'control']['complaint']
b = df[df['group'] == 'treatment']['complaint']

p_a, p_b, diff, z, p_complaint = proportion_z_test(a.sum(), len(a), b.sum(), len(b))

print(f'Control complaint rate: {p_a:.4f}')
print(f'Treatment complaint rate: {p_b:.4f}')
print(f'Absolute diff: {diff:.4f}')
print(f'p-value: {p_complaint:.6f}')
```

## 18. Поправка на множественные сравнения

Мы проверяем несколько метрик, поэтому растет риск случайно найти значимость. Использую поправку Holm: она менее жесткая, чем Бонферрони, но тоже контролирует ошибки.

```
# Собираю p-value по основным тестам.
tests = pd.DataFrame({
    'metric': ['watch_rate', 'watch_time', 'return_rate', 'revenue', 'complaint_rate'],
    'p_value': [p_watch, p_time, p_return, p_revenue, p_complaint]
})

# Поправка Holm.
tests = tests.sort_values('p_value').reset_index(drop=True)
m = len(tests)
tests['holm_threshold'] = 0.05 / (m - tests.index)
tests['significant_holm'] = tests['p_value'] <= tests['holm_threshold']
tests
```

**Короткий вывод:** после поправки стоит доверять только тем метрикам, которые остались значимыми. Это защищает от случайных выводов при большом числе проверок.

## 19. Сегментный анализ

Смотрю эффект главной метрики по устройствам и типам пользователей. Это помогает понять, где новая лента работает лучше. Но такие выводы нужно воспринимать осторожно, потому что сегментов много.

```
# Функция для сегментного анализа watch_rate.
def segment_test(col):
    rows = []
    for value in sorted(df[col].unique()):
        part = df[df[col] == value]
        a = part[part['group'] == 'control']['watched_video']
        b = part[part['group'] == 'treatment']['watched_video']
        p_a, p_b, diff, z, p = proportion_z_test(a.sum(), len(a), b.sum(), len(b))
        rows.append({'segment': value, 'control_rate': p_a, 'treatment_rate': p_b, 'uplift': diff / p_a, 'p_value': p, 'users': len(part)})
    return pd.DataFrame(rows).sort_values('uplift', ascending=False)

# Эффект по устройствам.
segment_test('device')
```

```
# Эффект по типу пользователя.
segment_test('user_type')
```

## 20. Power analysis

Оцениваю, сколько пользователей нужно на группу, чтобы заметить эффект в конверсии. Это помогает планировать следующий эксперимент.

```
# Простая оценка размера выборки для двух долей.
alpha = 0.05
power = 0.8
p_base = df[df['group'] == 'control']['watched_video'].mean()
min_effect = 0.015

z_alpha = stats.norm.ppf(1 - alpha / 2)
z_power = stats.norm.ppf(power)
p2 = p_base + min_effect
p_avg = (p_base + p2) / 2

n_needed = ((z_alpha * np.sqrt(2 * p_avg * (1 - p_avg)) + z_power * np.sqrt(p_base * (1 - p_base) + p2 * (1 - p2))) ** 2) / (min_effect ** 2)

print(f'Базовая конверсия: {p_base:.4f}')
print(f'Минимальный эффект: {min_effect:.4f}')
print(f'Нужно пользователей на группу: {np.ceil(n_needed):.0f}')
print(f'Фактически пользователей на группу примерно: {len(df) / 2:.0f}')
```

## 21. Финальное решение

Собираю главные результаты в одну таблицу и формулирую решение для бизнеса.

```
# Таблица главных результатов.
final = pd.DataFrame({
    'metric': ['watch_rate', 'watch_time', 'return_rate', 'revenue', 'complaint_rate'],
    'control': [summary.loc['control', 'watch_rate'], summary.loc['control', 'avg_watch_time'], summary.loc['control', 'return_rate'], summary.loc['control', 'arpu'], summary.loc['control', 'complaint_rate']],
    'treatment': [summary.loc['treatment', 'watch_rate'], summary.loc['treatment', 'avg_watch_time'], summary.loc['treatment', 'return_rate'], summary.loc['treatment', 'arpu'], summary.loc['treatment', 'complaint_rate']],
    'p_value': [p_watch, p_time, p_return, p_revenue, p_complaint]
})

final['absolute_diff'] = final['treatment'] - final['control']
final['relative_uplift'] = final['absolute_diff'] / final['control']
final.round(4)
```

```
# Простая логика решения.
main_ok = p_watch < 0.05 and final.loc[final['metric'] == 'watch_rate', 'absolute_diff'].iloc[0] > 0
guardrail_bad = p_complaint < 0.05 and final.loc[final['metric'] == 'complaint_rate', 'absolute_diff'].iloc[0] > 0

if main_ok and not guardrail_bad:
    decision = 'Рекомендация: раскатывать новую рекомендательную ленту.'
elif main_ok and guardrail_bad:
    decision = 'Рекомендация: не раскатывать полностью. Есть рост жалоб, нужно доработать механику.'
else:
    decision = 'Рекомендация: не раскатывать. Главная метрика не показала надежного улучшения.'

print(decision)
```
