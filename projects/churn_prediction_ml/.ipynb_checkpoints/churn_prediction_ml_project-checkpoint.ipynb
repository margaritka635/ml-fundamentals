# ML pet project: прогноз оттока пользователей

**Тема:** подписочный онлайн-сервис

В этом проекте я строю модель, которая предсказывает, уйдет ли пользователь из сервиса в ближайший месяц. Это классическая задача бинарной классификации для бизнеса: если заранее найти пользователей с высоким риском оттока, им можно предложить удерживающую коммуникацию.

## План проекта

1. Сгенерировать реалистичные данные пользователей.
2. Провести EDA и понять, кто чаще уходит.
3. Подготовить признаки для модели.
4. Обучить несколько моделей.
5. Сравнить качество по ROC-AUC, F1, recall и precision.
6. Выбрать рабочий порог классификации.
7. Посмотреть важность признаков.
8. Сделать бизнес-выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, моделей и метрик.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.metrics import confusion_matrix, classification_report, roc_curve
np.random.seed(42)
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Генерация данных

Данные синтетические, но логика похожа на реальный подписочный продукт. У пользователя есть тариф, активность, платежи, обращения в поддержку и история использования сервиса.

```
n = 12000
user_id = np.arange(1, n + 1)
age = np.random.randint(18, 65, size=n)
city_tier = np.random.choice(['large', 'medium', 'small'], size=n, p=[0.48, 0.34, 0.18])
plan = np.random.choice(['basic', 'standard', 'premium'], size=n, p=[0.46, 0.38, 0.16])
tenure_months = np.random.gamma(shape=2.2, scale=8, size=n).astype(int) + 1
monthly_fee = np.select([plan == 'basic', plan == 'standard', plan == 'premium'], [399, 699, 1099])
monthly_fee = monthly_fee + np.random.normal(0, 35, size=n)
watch_hours = np.random.gamma(shape=2.0, scale=8.0, size=n)
watch_hours += np.where(plan == 'premium', 8, 0)
watch_hours += np.where(plan == 'standard', 3, 0)
days_since_last_visit = np.random.exponential(scale=7, size=n).astype(int)
support_tickets = np.random.poisson(lam=0.35, size=n)
failed_payments = np.random.poisson(lam=0.18, size=n)
used_promo = np.random.binomial(1, 0.28, size=n)
auto_payment = np.random.binomial(1, 0.62, size=n)
family_account = np.random.binomial(1, np.where(plan == 'premium', 0.42, 0.18))
base = -2.1
base += 0.055 * days_since_last_visit
base += 0.35 * failed_payments
base += 0.18 * support_tickets
base += -0.035 * watch_hours
base += -0.025 * tenure_months
base += np.where(plan == 'basic', 0.35, 0)
base += np.where(plan == 'premium', -0.25, 0)
base += np.where(auto_payment == 1, -0.45, 0)
base += np.where(used_promo == 1, 0.18, 0)
base += np.where(family_account == 1, -0.22, 0)
prob = 1 / (1 + np.exp(-base))
churn = np.random.binomial(1, np.clip(prob, 0.01, 0.95))
df = pd.DataFrame({
    'user_id': user_id,
    'age': age,
    'city_tier': city_tier,
    'plan': plan,
    'tenure_months': tenure_months,
    'monthly_fee': monthly_fee.round(2),
    'watch_hours': watch_hours.round(2),
    'days_since_last_visit': days_since_last_visit,
    'support_tickets': support_tickets,
    'failed_payments': failed_payments,
    'used_promo': used_promo,
    'auto_payment': auto_payment,
    'family_account': family_account,
    'churn': churn
})
print(f'Размер данных: {df.shape[0]} строк, {df.shape[1]} колонок')
df.head()
```

**Короткий вывод:** данные содержат признаки поведения, платежей и профиля пользователя. Целевая переменная churn показывает, ушел пользователь или нет.

## 3. Проверка данных

Сначала проверяю пропуски, дубли и долю оттока. Это базовая проверка перед любой моделью.

```
print('Пропусков:', df.isna().sum().sum())
print('Дублей user_id:', df['user_id'].duplicated().sum())
print('Доля оттока:', round(df['churn'].mean(), 3))
```

```
sns.countplot(data=df, x='churn', color='steelblue')
plt.title('Баланс классов')
plt.xlabel('Churn')
plt.ylabel('Количество пользователей')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** если отток заметно меньше 50 процентов, accuracy может быть обманчивой. Поэтому дальше смотрю ROC-AUC, recall, precision и F1.

## 4. EDA по важным признакам

Смотрю, как отток связан с активностью, платежами и тарифом.

```
df.groupby('churn')[['watch_hours', 'days_since_last_visit', 'tenure_months', 'failed_payments', 'support_tickets']].mean().round(2)
```

```
sns.boxplot(data=df, x='churn', y='days_since_last_visit', color='steelblue')
plt.title('Дни с последнего визита и отток')
plt.xlabel('Churn')
plt.ylabel('Дней с последнего визита')
plt.tight_layout()
plt.show()
```

```
sns.boxplot(data=df, x='churn', y='watch_hours', color='salmon')
plt.title('Часы просмотра и отток')
plt.xlabel('Churn')
plt.ylabel('Часы просмотра')
plt.tight_layout()
plt.show()
```

```
plan_churn = df.groupby('plan')['churn'].mean().sort_values(ascending=False).reset_index()
sns.barplot(data=plan_churn, x='plan', y='churn', color='steelblue')
plt.title('Доля оттока по тарифу')
plt.xlabel('Тариф')
plt.ylabel('Доля оттока')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** чаще уходят пользователи с низкой активностью, большим числом дней с последнего визита и проблемами с платежами. Это похоже на реальную бизнес-логику.

## 5. Feature engineering

Создаю несколько простых признаков, которые могут помочь модели: активность на месяц подписки, флаг долгого отсутствия и флаг проблем с платежом.

```
df['hours_per_month'] = df['watch_hours'] / df['tenure_months']
df['long_absence'] = (df['days_since_last_visit'] >= 14).astype(int)
df['has_payment_problem'] = (df['failed_payments'] > 0).astype(int)
df[['hours_per_month', 'long_absence', 'has_payment_problem']].head()
```

## 6. Подготовка данных для модели

Убираю user_id, потому что это технический идентификатор. Категориальные признаки кодирую через one-hot encoding.

```
y = df['churn']
x = df.drop(columns=['user_id', 'churn'])
x = pd.get_dummies(x, drop_first=True)
print(f'x: {x.shape}')
print(f'y: {y.shape}')
```

## 7. Train и validation

Делю данные на train и validation. Stratify сохраняет похожую долю оттока в обеих частях.

```
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
print('Доля churn в train:', y_train.mean().round(3))
print('Доля churn в val:  ', y_val.mean().round(3))
```

## 8. Масштабирование для логистической регрессии

Логистической регрессии полезно масштабирование признаков. Для деревьев и бустинга масштабирование не нужно.

```
scaler = StandardScaler()
x_train_sc = scaler.fit_transform(x_train)
x_val_sc = scaler.transform(x_val)
```

## 9. Обучение моделей

Сравниваю простую интерпретируемую модель и две модели на деревьях.

```
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced'),
    'Random Forest': RandomForestClassifier(n_estimators=250, max_depth=7, min_samples_leaf=10, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=180, learning_rate=0.05, max_depth=3, random_state=42)
}
models
```

```
res = []
for name, model in models.items():
    if name == 'Logistic Regression':
        model.fit(x_train_sc, y_train)
        pred = model.predict(x_val_sc)
        prob = model.predict_proba(x_val_sc)[:, 1]
    else:
        model.fit(x_train, y_train)
        pred = model.predict(x_val)
        prob = model.predict_proba(x_val)[:, 1]
    res.append({'model': name, 'accuracy': accuracy_score(y_val, pred), 'precision': precision_score(y_val, pred), 'recall': recall_score(y_val, pred), 'f1': f1_score(y_val, pred), 'roc_auc': roc_auc_score(y_val, prob)})
res = pd.DataFrame(res).sort_values('roc_auc', ascending=False)
res
```

**Короткий вывод:** ROC-AUC показывает, насколько хорошо модель ранжирует пользователей по риску оттока. Recall важен, если бизнесу нужно найти как можно больше пользователей с риском ухода.

## 10. Выбор лучшей модели

Беру модель с максимальным ROC-AUC и смотрю подробный отчет по классам.

```
best_name = res.iloc[0]['model']
best_model = models[best_name]
if best_name == 'Logistic Regression':
    prob = best_model.predict_proba(x_val_sc)[:, 1]
else:
    prob = best_model.predict_proba(x_val)[:, 1]
pred = (prob >= 0.5).astype(int)
print(f'Лучшая модель: {best_name}')
print(classification_report(y_val, pred))
```

## 11. Матрица ошибок

Матрица ошибок показывает, сколько пользователей модель правильно и неправильно отнесла к оттоку.

```
cm = confusion_matrix(y_val, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title(f'Матрица ошибок: {best_name}')
plt.xlabel('Прогноз')
plt.ylabel('Факт')
plt.tight_layout()
plt.show()
```

## 12. ROC-кривая

ROC-кривая показывает качество модели при разных порогах. Чем выше кривая, тем лучше модель отделяет churn от не churn.

```
fpr, tpr, thresholds = roc_curve(y_val, prob)
auc = roc_auc_score(y_val, prob)
plt.plot(fpr, tpr, color='steelblue', label=f'ROC-AUC = {auc:.3f}')
plt.plot([0, 1], [0, 1], color='red', linestyle='--')
plt.title('ROC-кривая')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.tight_layout()
plt.show()
```

## 13. Подбор порога

Порог 0.5 не всегда лучший. Если бизнес хочет поймать больше пользователей с риском оттока, можно снизить порог и поднять recall.

```
rows = []
for threshold in np.arange(0.2, 0.81, 0.05):
    p = (prob >= threshold).astype(int)
    rows.append({'threshold': threshold, 'precision': precision_score(y_val, p), 'recall': recall_score(y_val, p), 'f1': f1_score(y_val, p)})
thr = pd.DataFrame(rows)
thr.head()
```

```
plt.plot(thr['threshold'], thr['precision'], marker='o', label='precision')
plt.plot(thr['threshold'], thr['recall'], marker='o', label='recall')
plt.plot(thr['threshold'], thr['f1'], marker='o', label='f1')
plt.title('Метрики при разных порогах')
plt.xlabel('Порог')
plt.ylabel('Метрика')
plt.legend()
plt.tight_layout()
plt.show()
```

**Короткий вывод:** порог выбирается не только по максимуму F1. Если удерживающая акция дешевая, можно выбрать порог ниже и ловить больше пользователей с риском оттока.

## 14. Важность признаков

Для моделей на деревьях можно посмотреть важность признаков. Это помогает объяснить, почему модель считает пользователя рискованным.

```
if hasattr(best_model, 'feature_importances_'):
    imp = pd.DataFrame({'feature': x_train.columns, 'importance': best_model.feature_importances_})
    imp = imp.sort_values('importance', ascending=False).head(15)
else:
    coef = best_model.coef_[0]
    imp = pd.DataFrame({'feature': x_train.columns, 'importance': np.abs(coef)})
    imp = imp.sort_values('importance', ascending=False).head(15)
imp
```

```
sns.barplot(data=imp, x='importance', y='feature', color='steelblue')
plt.title('Топ-15 важных признаков')
plt.xlabel('Важность')
plt.ylabel('Признак')
plt.tight_layout()
plt.show()
```

## 15. Сегментация пользователей по риску

Разделяю пользователей validation на группы риска. Это удобно для бизнес-действий: высокий риск получает удерживающую коммуникацию первым.

```
val_result = x_val.copy()
val_result['true_churn'] = y_val.values
val_result['churn_prob'] = prob
val_result['risk_group'] = pd.cut(val_result['churn_prob'], bins=[0, 0.3, 0.6, 1.0], labels=['low', 'medium', 'high'])
val_result.groupby('risk_group', observed=False).agg(users=('churn_prob', 'count'), avg_prob=('churn_prob', 'mean'), real_churn=('true_churn', 'mean')).round(3)
```
