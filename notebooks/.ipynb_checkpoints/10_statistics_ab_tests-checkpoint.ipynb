# Статистика и A/B тесты

**Датасет:** синтетические данные

В этом ноутбуке я разбираю базовые статистические методы для A/B-тестов. Данные искусственные, зато мы точно знаем, какой эффект заложен, и можем спокойно проверить методы.

## План работы

1. Сгенерировать данные для A/B-теста.
2. Посмотреть распределения групп.
3. Посчитать доверительный интервал для среднего.
4. Провести t-test для независимых выборок.
5. Проверить разницу через bootstrap.
6. Провести хи-квадрат тест для конверсии.
7. Оценить размер выборки.
8. Разобрать множественные сравнения и поправку Бонферрони.
9. Сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, расчетов, графиков и статистических тестов.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import ttest_ind, chi2_contingency
np.random.seed(42)
%matplotlib inline
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Генерация данных

Создаю две группы: control и treatment. В treatment среднее немного выше, как будто новая версия продукта улучшила метрику.

```python
# Размер каждой группы
n = 1000
# Метрика в контрольной группе
control = np.random.normal(loc=10.0, scale=2.0, size=n)
# Метрика в тестовой группе. Среднее чуть выше
treatment = np.random.normal(loc=10.5, scale=2.0, size=n)
# Собираю данные в одну таблицу
df = pd.DataFrame({
    'group': ['control'] * n + ['treatment'] * n,
    'value': np.concatenate([control, treatment])
})
print(f'Control mean: {control.mean():.3f}')
print(f'Treatment mean: {treatment.mean():.3f}')
print(f'Difference: {treatment.mean() - control.mean():.3f}')
df.head()
```

**Короткий вывод:** среднее в treatment выше. Но по одной разнице средних нельзя сразу делать вывод: нужно понять, могла ли такая разница появиться случайно.

## 3. Визуальный анализ

Смотрю распределения двух групп. Если группы сильно перекрываются, эффект может быть небольшим относительно шума.

```python
# Строю гистограммы двух групп
sns.histplot(data=df, x='value', hue='group', bins=40, kde=True, alpha=0.5)
plt.title('Распределение метрики в A/B группах')
plt.xlabel('Значение метрики')
plt.ylabel('Количество')
plt.tight_layout()
plt.show()
```

```python
# Считаю основные статистики по группам
df.groupby('group')['value'].agg(['count', 'mean', 'std', 'median']).round(3)
```

## 4. Доверительный интервал для среднего

Доверительный интервал показывает диапазон, где с высокой вероятностью находится истинное среднее группы.

```python
# Функция для доверительного интервала среднего
def mean_ci(x, alpha=0.05):
    x = np.array(x)
    mean = x.mean()
    se = stats.sem(x)
    t_value = stats.t.ppf(1 - alpha / 2, len(x) - 1)
    left = mean - t_value * se
    right = mean + t_value * se
    return mean, left, right
# Считаю интервалы для обеих групп
control_mean, control_left, control_right = mean_ci(control)
treat_mean, treat_left, treat_right = mean_ci(treatment)
print(f'Control mean: {control_mean:.3f}, CI: [{control_left:.3f}, {control_right:.3f}]')
print(f'Treatment mean: {treat_mean:.3f}, CI: [{treat_left:.3f}, {treat_right:.3f}]')
```

**Короткий вывод:** доверительные интервалы помогают увидеть неопределенность вокруг среднего. Если интервалы почти не пересекаются, это уже намек на реальную разницу.

## 5. t-test для независимых выборок

t-test проверяет гипотезу, что средние двух групп равны. Если p-value меньше 0.05, разницу обычно считают статистически значимой.

```python
# Запускаю t-test. equal_var=False - более аккуратный вариант, если дисперсии групп отличаются
t_stat, p_value = ttest_ind(control, treatment, equal_var=False)
print(f't-statistic: {t_stat:.4f}')
print(f'p-value: {p_value:.6f}')
if p_value < 0.05:
    print('Вывод: разница статистически значима.')
else:
    print('Вывод: статистически значимой разницы нет.')
```

## 6. Размер эффекта

Статистическая значимость не говорит, насколько эффект большой. Поэтому считаю Cohen d: разницу средних в единицах общего стандартного отклонения.

```python
# Считаю Cohen d
pooled_std = np.sqrt((control.var(ddof=1) + treatment.var(ddof=1)) / 2)
cohen_d = (treatment.mean() - control.mean()) / pooled_std
print(f'Cohen d: {cohen_d:.3f}')
```

**Короткий вывод:** даже если p-value маленький, эффект может быть умеренным. Для бизнеса важно смотреть и значимость, и размер эффекта.

## 7. Bootstrap для разницы средних

Bootstrap много раз пересэмплирует данные и строит распределение разницы средних. Это простой способ получить доверительный интервал без строгих формул.

```python
# Количество bootstrap-итераций
b = 5000
# Здесь будут храниться разницы средних
diffs = []
# Много раз берем выборки с возвращением и считаем разницу средних
for i in range(b):
    c_sample = np.random.choice(control, size=len(control), replace=True)
    t_sample = np.random.choice(treatment, size=len(treatment), replace=True)
    diffs.append(t_sample.mean() - c_sample.mean())
diffs = np.array(diffs)
# Считаю 95 процентный доверительный интервал
left = np.percentile(diffs, 2.5)
right = np.percentile(diffs, 97.5)
print(f'Bootstrap mean diff: {diffs.mean():.3f}')
print(f'95% CI: [{left:.3f}, {right:.3f}]')
```

```python
# Строю распределение bootstrap-разниц
sns.histplot(diffs, bins=40, color='steelblue')
plt.axvline(0, color='red', linestyle='--', label='Нет эффекта')
plt.axvline(left, color='black', linestyle='--', label='95% CI')
plt.axvline(right, color='black', linestyle='--')
plt.title('Bootstrap распределение разницы средних')
plt.xlabel('Treatment mean минус control mean')
plt.ylabel('Количество')
plt.legend()
plt.tight_layout()
plt.show()
```

**Короткий вывод:** если доверительный интервал разницы не включает 0, это поддерживает вывод о наличии эффекта.

## 8. Хи-квадрат тест для конверсии

Теперь создаю пример с бинарной метрикой: конверсия есть или нет. Для такой задачи подходит хи-квадрат тест.

```python
# Генерирую конверсии для двух групп
np.random.seed(42)
control_conv = np.random.binomial(1, 0.10, size=n)
treat_conv = np.random.binomial(1, 0.12, size=n)
# Собираю таблицу сопряженности
conv_table = pd.DataFrame({
    'converted': [control_conv.sum(), treat_conv.sum()],
    'not_converted': [n - control_conv.sum(), n - treat_conv.sum()]
}, index=['control', 'treatment'])
conv_table
```

```python
# Запускаю хи-квадрат тест
chi2, p_value, dof, expected = chi2_contingency(conv_table)
print(f'Control conversion: {control_conv.mean():.3f}')
print(f'Treatment conversion: {treat_conv.mean():.3f}')
print(f'chi2: {chi2:.4f}')
print(f'p-value: {p_value:.6f}')
if p_value < 0.05:
    print('Вывод: разница в конверсии статистически значима.')
else:
    print('Вывод: статистически значимой разницы в конверсии нет.')
```

## 9. Оценка размера выборки

Чем меньше эффект, тем больше нужна выборка. Ниже простая оценка размера группы для сравнения двух средних.

```python
# Задаю параметры для примерной оценки размера выборки
alpha = 0.05
power = 0.8
sigma = 2.0
min_effect = 0.3
# z-значения для alpha и power
z_alpha = stats.norm.ppf(1 - alpha / 2)
z_power = stats.norm.ppf(power)
# Простая формула для двух независимых групп одинакового размера
n_needed = 2 * ((z_alpha + z_power) * sigma / min_effect) ** 2
print(f'Минимальный эффект: {min_effect}')
print(f'Примерный размер на группу: {np.ceil(n_needed):.0f}')
```

**Короткий вывод:** маленькие эффекты требуют больших выборок. Если данных мало, тест может не увидеть даже реальное улучшение.

## 10. Множественные сравнения

Если проверять много гипотез, растет шанс случайно найти значимый результат. Поправка Бонферрони делает порог значимости строже.

```python
# Генерирую несколько метрик для примера
np.random.seed(42)
metrics = []
for i in range(5):
    c = np.random.normal(10, 2, n)
    t = np.random.normal(10 + 0.15 * i, 2, n)
    p = ttest_ind(c, t, equal_var=False).pvalue
    metrics.append({'metric': f'metric_{i + 1}', 'p_value': p})
multi = pd.DataFrame(metrics)
# Поправка Бонферрони: p-value умножается на число тестов
m = len(multi)
multi['p_bonferroni'] = (multi['p_value'] * m).clip(upper=1)
multi['significant_before'] = multi['p_value'] < 0.05
multi['significant_after'] = multi['p_bonferroni'] < 0.05
multi
```

**Короткий вывод:** после поправки часть значимых результатов может исчезнуть. Это нормально: мы защищаемся от случайных находок при большом числе тестов.

## Итоги

В этом ноутбуке я разобрала базовые инструменты статистики и A/B-тестов.

- Доверительный интервал показывает неопределенность вокруг оценки.
- t-test проверяет разницу средних между двумя независимыми группами.
- Cohen d помогает понять размер эффекта.
- Bootstrap дает доверительный интервал через пересэмплирование.
- Хи-квадрат тест подходит для сравнения конверсий.
- Размер выборки зависит от ожидаемого эффекта, шума, alpha и мощности.
- При множественных сравнениях нужна поправка, например Бонферрони.

Главный вывод: в A/B-тестах важно смотреть не только на p-value. Нужно учитывать размер эффекта, доверительный интервал, мощность теста и бизнес-смысл изменения.
