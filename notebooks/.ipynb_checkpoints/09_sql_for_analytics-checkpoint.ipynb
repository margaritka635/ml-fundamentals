# SQL для аналитики

**Датасет:** House Prices

В этом ноутбуке я использую SQLite для аналитики по продажам домов. Данные загружаются из csv в базу, а дальше основные вопросы решаются SQL-запросами.

## План работы

1. Загрузить данные и создать SQLite базу.
2. Разделить данные на таблицы sales и houses.
3. Сделать простые SELECT-запросы с фильтрами и сортировкой.
4. Использовать GROUP BY и HAVING.
5. Показать JOIN между таблицами.
6. Использовать CTE и подзапросы.
7. Показать оконные функции ROW_NUMBER, RANK и LAG.
8. Сделать итоговые выводы.

## 1. Импорт библиотек

Подключаю pandas для работы с таблицами и sqlite3 для создания локальной базы данных прямо в памяти.

```python
import pandas as pd
import sqlite3
conn = sqlite3.connect(':memory:')
```

## 2. Загрузка данных

Загружаю данные о домах. Для SQL-аналитики удобнее разделить одну большую таблицу на несколько логических таблиц.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Проверяю размер таблицы
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонка')
df.head()
```

## 3. Создание таблиц

Создаю две таблицы. Sales хранит информацию о продаже и цене. Houses хранит основные характеристики дома. Так можно показать JOIN как в реальной аналитической базе.

```python
# Таблица продаж
sales = df[['Id', 'SalePrice', 'YrSold', 'MoSold', 'SaleCondition']].copy()
# Таблица характеристик домов
houses = df[['Id', 'Neighborhood', 'BldgType', 'HouseStyle', 'OverallQual', 'GrLivArea', 'TotalBsmtSF', 'GarageCars', 'YearBuilt']].copy()
# Записываю таблицы в SQLite
sales.to_sql('sales', conn, if_exists='replace', index=False)
houses.to_sql('houses', conn, if_exists='replace', index=False)
# Проверяю количество строк в таблицах
print(pd.read_sql('SELECT COUNT(*) AS n FROM sales', conn))
print(pd.read_sql('SELECT COUNT(*) AS n FROM houses', conn))
```

**Короткий вывод:** в базе есть две таблицы с одинаковым ключом Id. Теперь можно писать SQL-запросы и объединять данные.

## 4. Простые SELECT, WHERE и ORDER BY

Сначала выбираю самые дорогие дома. WHERE фильтрует строки, ORDER BY сортирует результат.

```python
# Топ-10 самых дорогих домов
query = """
SELECT Id, SalePrice, YrSold, MoSold, SaleCondition
FROM sales
WHERE SalePrice > 300000
ORDER BY SalePrice DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

**Короткий вывод:** простой запрос уже позволяет быстро найти дорогие сделки и посмотреть, в какие месяцы и годы они были.

## 5. Агрегации GROUP BY

Считаю среднюю цену и количество продаж по годам. GROUP BY собирает строки в группы, а агрегатные функции считают показатели внутри группы.

```python
# Средняя цена и количество продаж по годам
query = """
SELECT
    YrSold,
    COUNT(*) AS sales_count,
    ROUND(AVG(SalePrice), 0) AS avg_price,
    MIN(SalePrice) AS min_price,
    MAX(SalePrice) AS max_price
FROM sales
GROUP BY YrSold
ORDER BY YrSold
"""
pd.read_sql(query, conn)
```

## 6. GROUP BY и HAVING

HAVING фильтрует уже готовые группы. Здесь оставляю только районы, где было не меньше 50 продаж.

```python
# Районы с достаточным числом продаж
query = """
SELECT
    h.Neighborhood,
    COUNT(*) AS sales_count,
    ROUND(AVG(s.SalePrice), 0) AS avg_price
FROM sales AS s
JOIN houses AS h ON s.Id = h.Id
GROUP BY h.Neighborhood
HAVING COUNT(*) >= 50
ORDER BY avg_price DESC
"""
pd.read_sql(query, conn)
```

**Короткий вывод:** HAVING помогает не делать выводы по слишком маленьким группам. Район с одной дорогой продажей не должен автоматически считаться самым дорогим.

## 7. JOIN таблиц

JOIN соединяет продажи с характеристиками домов. Так можно анализировать цену вместе с районом, площадью и качеством дома.

```python
# Соединяю продажи и признаки домов
query = """
SELECT
    s.Id,
    s.SalePrice,
    h.Neighborhood,
    h.OverallQual,
    h.GrLivArea,
    h.GarageCars
FROM sales AS s
JOIN houses AS h ON s.Id = h.Id
ORDER BY s.SalePrice DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

## 8. CTE

CTE помогает разбить сложный запрос на понятные шаги. Здесь сначала считаю среднюю цену по району, а потом выбираю районы дороже общей средней цены.

```python
# Районы, где средняя цена выше общей средней
query = """
WITH neighborhood_prices AS (
    SELECT h.Neighborhood, COUNT(*) AS sales_count, AVG(s.SalePrice) AS avg_price
    FROM sales AS s
    JOIN houses AS h ON s.Id = h.Id
    GROUP BY h.Neighborhood
), total_price AS (
    SELECT AVG(SalePrice) AS total_avg_price
    FROM sales
)
SELECT
    n.Neighborhood,
    n.sales_count,
    ROUND(n.avg_price, 0) AS avg_price,
    ROUND(t.total_avg_price, 0) AS total_avg_price
FROM neighborhood_prices AS n
CROSS JOIN total_price AS t
WHERE n.avg_price > t.total_avg_price
ORDER BY n.avg_price DESC
"""
pd.read_sql(query, conn)
```

**Короткий вывод:** CTE делает запрос читаемым: сначала считаются промежуточные таблицы, потом из них собирается итоговый результат.

## 9. Подзапрос

Подзапрос можно использовать прямо внутри WHERE. Здесь выбираю дома дороже средней цены по всему датасету.

```python
# Дома дороже средней цены
query = """
SELECT
    s.Id,
    s.SalePrice,
    h.Neighborhood,
    h.OverallQual
FROM sales AS s
JOIN houses AS h ON s.Id = h.Id
WHERE s.SalePrice > (SELECT AVG(SalePrice) FROM sales)
ORDER BY s.SalePrice DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

## 10. Оконная функция ROW_NUMBER

ROW_NUMBER нумерует строки внутри группы. Так можно найти самый дорогой дом в каждом районе.

```python
# Самый дорогой дом в каждом районе
query = """
WITH ranked AS (
    SELECT
        h.Neighborhood,
        s.Id,
        s.SalePrice,
        h.OverallQual,
        ROW_NUMBER() OVER (PARTITION BY h.Neighborhood ORDER BY s.SalePrice DESC) AS rn
    FROM sales AS s
    JOIN houses AS h ON s.Id = h.Id
)
SELECT Neighborhood, Id, SalePrice, OverallQual
FROM ranked
WHERE rn = 1
ORDER BY SalePrice DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

## 11. Оконная функция RANK

RANK показывает место объекта внутри группы. Здесь я ранжирую дома внутри каждого года продажи по цене.

```python
# Рейтинг сделок по цене внутри каждого года
query = """
SELECT
    Id,
    YrSold,
    SalePrice,
    RANK() OVER (PARTITION BY YrSold ORDER BY SalePrice DESC) AS price_rank
FROM sales
ORDER BY YrSold, price_rank
LIMIT 15
"""
pd.read_sql(query, conn)
```

## 12. Оконная функция LAG

LAG берет значение из предыдущей строки. Здесь сравниваю среднюю цену месяца с предыдущим месяцем.

```python
# Изменение средней цены от месяца к месяцу
query = """
WITH monthly AS (
    SELECT YrSold, MoSold, AVG(SalePrice) AS avg_price
    FROM sales
    GROUP BY YrSold, MoSold
), with_lag AS (
    SELECT YrSold, MoSold, avg_price, LAG(avg_price) OVER (ORDER BY YrSold, MoSold) AS prev_avg_price
    FROM monthly
)
SELECT
    YrSold,
    MoSold,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(prev_avg_price, 0) AS prev_avg_price,
    ROUND(avg_price - prev_avg_price, 0) AS diff
FROM with_lag
ORDER BY YrSold, MoSold
LIMIT 15
"""
pd.read_sql(query, conn)
```

**Короткий вывод:** оконные функции полезны, когда нужно сравнивать строки между собой: место в рейтинге, предыдущий месяц, топ внутри группы.

## 13. Практический аналитический запрос

В одном запросе собираю мини-отчет по районам: количество продаж, среднюю цену, среднее качество и место района по средней цене.

```python
# Мини-отчет по районам
query = """
WITH report AS (
    SELECT
        h.Neighborhood,
        COUNT(*) AS sales_count,
        AVG(s.SalePrice) AS avg_price,
        AVG(h.OverallQual) AS avg_quality,
        AVG(h.GrLivArea) AS avg_area
    FROM sales AS s
    JOIN houses AS h ON s.Id = h.Id
    GROUP BY h.Neighborhood
)
SELECT
    Neighborhood,
    sales_count,
    ROUND(avg_price, 0) AS avg_price,
    ROUND(avg_quality, 2) AS avg_quality,
    ROUND(avg_area, 0) AS avg_area,
    RANK() OVER (ORDER BY avg_price DESC) AS price_rank
FROM report
ORDER BY price_rank
LIMIT 15
"""
pd.read_sql(query, conn)
```
