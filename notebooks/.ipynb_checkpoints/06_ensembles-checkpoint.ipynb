# Ансамблевые методы

**Датасет:** House Prices

В этом ноутбуке я сравниваю ансамблевые модели на задаче классификации дорогих домов. Ансамбли объединяют много простых моделей и часто дают качество выше, чем одно дерево решений.

## План работы

1. Загрузить данные и создать бинарную цель.
2. Подготовить признаки для моделей.
3. Обучить Random Forest и Gradient Boosting.
4. Добавить XGBoost, CatBoost и LightGBM, если библиотеки доступны.
5. Сравнить модели по ROC-AUC и F1.
6. Посмотреть важность признаков у лучшей модели.
7. Сделать выводы.

## 1. Импорт библиотек

Подключаю основные библиотеки, модели ансамблей и метрики классификации. Внешние библиотеки подключаю через try, чтобы ноутбук не падал, если какая-то из них не установлена.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.metrics import classification_report, confusion_matrix
try:
    from xgboost import XGBClassifier
    has_xgb = True
except Exception:
    has_xgb = False
try:
    from catboost import CatBoostClassifier
    has_cat = True
except Exception:
    has_cat = False
try:
    from lightgbm import LGBMClassifier
    has_lgb = True
except Exception:
    has_lgb = False
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
print('XGBoost доступен:', has_xgb)
print('CatBoost доступен:', has_cat)
print('LightGBM доступен:', has_lgb)
```

## 2. Загрузка данных

Загружаю данные о домах и создаю бинарную цель. Дорогим считаю дом, цена которого выше 75 процентиля.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Создаю границу дорогого дома
limit = df['SalePrice'].quantile(0.75)
# Создаю бинарную цель
df['HighPrice'] = (df['SalePrice'] > limit).astype(int)
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонки')
print(f'Граница дорогого дома: {limit:.0f}')
df.head()
```

**Короткий вывод:** задача такая же, как в деревьях решений: нужно отделить дорогие дома от остальных.

## 3. Баланс классов

Проверяю распределение классов. Если классы не равны, важно смотреть не только accuracy, но и F1, recall и ROC-AUC.

```python
# Считаю количество объектов каждого класса
class_count = df['HighPrice'].value_counts().sort_index()
class_count
```

```python
# Строю график баланса классов
sns.countplot(data=df, x='HighPrice', color='steelblue')
plt.title('Баланс классов')
plt.xlabel('Класс')
plt.ylabel('Количество домов')
plt.tight_layout()
plt.show()
```

## 4. Подготовка признаков

Ансамбли на деревьях не требуют масштабирования. Но текстовые признаки нужно закодировать, потому что большинство моделей принимает только числовые признаки.

```python
# y - ответ: дорогой дом или нет
y = df['HighPrice']
# x - признаки. Id, SalePrice и HighPrice убираю
x = df.drop(columns=['Id', 'SalePrice', 'HighPrice'])
print(f'x: {x.shape}')
print(f'y: {y.shape}')
```

```python
# Разделяю признаки на числовые и текстовые
num_cols = x.select_dtypes(include='number').columns
cat_cols = x.select_dtypes(include='object').columns
print(f'Числовых колонок: {len(num_cols)}')
print(f'Текстовых колонок: {len(cat_cols)}')
```

```python
# Числовые пропуски заполняю медианой
x[num_cols] = x[num_cols].fillna(x[num_cols].median())
```

```python
# Текстовые пропуски заполняю словом None
x[cat_cols] = x[cat_cols].fillna('None')
```

```python
# Кодирую текстовые признаки
x = pd.get_dummies(x, drop_first=True)
print(f'Размер x после кодирования: {x.shape}')
print(f'Осталось пропусков: {x.isna().sum().sum()}')
```

## 5. Разделение на train и validation

Делю данные на обучение и проверку. Stratify сохраняет похожую долю дорогих домов в обеих частях.

```python
# 80 процентов данных идут на обучение, 20 процентов - на проверку
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
print('Доля класса 1 в train:', y_train.mean().round(3))
print('Доля класса 1 в val:  ', y_val.mean().round(3))
```

## 6. Список моделей

Собираю модели в один словарь. Random Forest использует много деревьев, а бустинги строят деревья последовательно и исправляют ошибки предыдущих.

```python
# Создаю базовый набор ансамблевых моделей
models = {
    'Random Forest': RandomForestClassifier(n_estimators=300, max_depth=6, min_samples_leaf=5, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=200, learning_rate=0.05, max_depth=3, random_state=42)
}
# Добавляю XGBoost, если он установлен
if has_xgb:
    models['XGBoost'] = XGBClassifier(n_estimators=200, max_depth=3, learning_rate=0.05, subsample=0.8, colsample_bytree=0.8, eval_metric='logloss', random_state=42, n_jobs=1)
# Добавляю CatBoost, если он установлен
if has_cat:
    models['CatBoost'] = CatBoostClassifier(iterations=200, depth=4, learning_rate=0.05, loss_function='Logloss', verbose=False, random_state=42, allow_writing_files=False)
# Добавляю LightGBM, если он установлен
if has_lgb:
    models['LightGBM'] = LGBMClassifier(n_estimators=200, learning_rate=0.05, max_depth=4, random_state=42)
models
```

## 7. Обучение и сравнение моделей

Обучаю все доступные модели одинаковым способом и сравниваю качество на validation.

```python
# В эту таблицу буду складывать метрики
res = []
# Обучаю модели и считаю качество
for name, model in models.items():
    model.fit(x_train, y_train)
    pred = model.predict(x_val)
    prob = model.predict_proba(x_val)[:, 1]
    res.append({'model': name, 'accuracy': accuracy_score(y_val, pred), 'precision': precision_score(y_val, pred), 'recall': recall_score(y_val, pred), 'f1': f1_score(y_val, pred), 'roc_auc': roc_auc_score(y_val, prob)})
# Собираю результаты в таблицу и сортирую по ROC-AUC
res = pd.DataFrame(res).sort_values('roc_auc', ascending=False)
res
```

**Короткий вывод:** ROC-AUC показывает, насколько хорошо модель ранжирует дома по вероятности быть дорогими. F1 помогает оценить баланс между precision и recall.

## 8. График сравнения

Строю график ROC-AUC для всех моделей, чтобы быстро увидеть лидера.

```python
# Строю график ROC-AUC
sns.barplot(data=res, x='roc_auc', y='model', color='steelblue')
plt.title('Сравнение ансамблей по ROC-AUC')
plt.xlabel('ROC-AUC')
plt.ylabel('Модель')
plt.tight_layout()
plt.show()
```

## 9. Подробная оценка лучшей модели

Беру модель с максимальным ROC-AUC и смотрю отчет по классам.

```python
# Выбираю лучшую модель
best_name = res.iloc[0]['model']
best_model = models[best_name]
# Делаю прогноз лучшей модели
pred = best_model.predict(x_val)
prob = best_model.predict_proba(x_val)[:, 1]
print(f'Лучшая модель: {best_name}')
print(f'ROC-AUC: {roc_auc_score(y_val, prob):.4f}')
print(classification_report(y_val, pred))
```

## 10. Матрица ошибок

Матрица ошибок показывает, где модель ошибается: принимает обычные дома за дорогие или пропускает дорогие дома.

```python
# Строю матрицу ошибок
cm = confusion_matrix(y_val, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title(f'Матрица ошибок: {best_name}')
plt.xlabel('Прогноз')
plt.ylabel('Реальный класс')
plt.tight_layout()
plt.show()
```

## 11. Важность признаков

У большинства ансамблей на деревьях есть feature_importances_. Это помогает понять, какие признаки сильнее всего влияли на решение модели.

```python
# Достаю важности признаков, если модель их поддерживает
if hasattr(best_model, 'feature_importances_'):
    imp = pd.DataFrame({'feature': x_train.columns, 'importance': best_model.feature_importances_})
    imp = imp.sort_values('importance', ascending=False).head(15)
else:
    imp = pd.DataFrame(columns=['feature', 'importance'])
imp
```

```python
# Строю график важности признаков
if len(imp) > 0:
    sns.barplot(data=imp, x='importance', y='feature', color='steelblue')
    plt.title(f'Топ-15 важных признаков: {best_name}')
    plt.xlabel('Важность')
    plt.ylabel('Признак')
    plt.tight_layout()
    plt.show()
else:
    print('У лучшей модели нет feature_importances_')
```

**Короткий вывод:** важность признаков обычно показывает знакомые факторы: качество дома, площадь, гараж, подвал и качество кухни. Это делает модель не только точной, но и объяснимой.

## Итоги

В этом ноутбуке я сравнила несколько ансамблевых моделей для классификации дорогих домов.

- Random Forest строит много независимых деревьев и усредняет их ответы.
- Gradient Boosting строит деревья последовательно, исправляя ошибки предыдущих.
- XGBoost и CatBoost - более сильные реализации бустинга.
- LightGBM добавляется только если библиотека установлена в окружении.
- Для ансамблей на деревьях масштабирование признаков не требуется.
- Сравнение моделей лучше делать по нескольким метрикам: ROC-AUC, F1, precision и recall.

Главный вывод: ансамбли обычно сильнее одного дерева, потому что уменьшают случайность и лучше ловят сложные зависимости в данных.
