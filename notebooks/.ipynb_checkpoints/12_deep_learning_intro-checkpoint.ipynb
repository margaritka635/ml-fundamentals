# Введение в глубокое обучение

**Датасет:** Digits из sklearn

В этом ноутбуке я обучаю простую нейронную сеть на изображениях цифр. Данные локальные, поэтому ничего не нужно скачивать из интернета.

## План работы

1. Загрузить датасет с цифрами.
2. Подготовить train и validation.
3. Создать TensorDataset и DataLoader.
4. Описать простую нейронную сеть.
5. Обучить модель несколько эпох.
6. Построить графики loss и accuracy.
7. Посмотреть ошибки модели.
8. Сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, данных, PyTorch-модели и оценки качества.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
np.random.seed(42)
torch.manual_seed(42)
device = torch.device('cpu')
%matplotlib inline
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
print(f'Устройство: {device}')
```

## 2. Загрузка данных

Использую встроенный датасет digits. Каждое изображение - это цифра размером 8 на 8 пикселей. Всего 10 классов: цифры от 0 до 9.

```python
# Загружаю данные
data = load_digits()
# x - изображения в виде 64 признаков, y - правильные цифры
x = data.data
y = data.target
print(f'x: {x.shape}')
print(f'y: {y.shape}')
print(f'Классы: {data.target_names}')
```

## 3. Примеры изображений

Перед обучением смотрю на картинки. Так проще понимать, какую задачу решает модель.

```python
# Показываю первые 10 изображений
fig, ax = plt.subplots(2, 5, figsize=(10, 4))
for i, a in enumerate(ax.flatten()):
    a.imshow(data.images[i], cmap='gray')
    a.set_title(f'Класс: {y[i]}')
    a.axis('off')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** изображения маленькие, но цифры различимы. Для первой нейронной сети этого датасета достаточно.

## 4. Подготовка данных

Нормирую значения пикселей: исходно они от 0 до 16, а для нейронной сети удобнее диапазон примерно от 0 до 1.

```python
# Нормирую пиксели
x = x / 16.0
# Разделяю данные на train и validation с сохранением баланса классов
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
```

## 5. TensorDataset и DataLoader

PyTorch обучает модель батчами. TensorDataset хранит признаки и ответы, а DataLoader выдает данные небольшими порциями.

```python
# Перевожу numpy-массивы в torch-тензоры
x_train_t = torch.tensor(x_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.long)
x_val_t = torch.tensor(x_val, dtype=torch.float32)
y_val_t = torch.tensor(y_val, dtype=torch.long)
# Создаю датасеты
train_ds = TensorDataset(x_train_t, y_train_t)
val_ds = TensorDataset(x_val_t, y_val_t)
# Создаю загрузчики данных
train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=64, shuffle=False)
print(f'Батчей train: {len(train_loader)}')
print(f'Батчей val:   {len(val_loader)}')
```

## 6. Архитектура модели

Создаю простую полносвязную нейронную сеть. На входе 64 признака, на выходе 10 чисел - по одному для каждого класса.

```python
# Простая нейронная сеть для классификации цифр
class DigitNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 10)
        )
    def forward(self, x):
        return self.net(x)
# Создаю модель и переношу ее на выбранное устройство
model = DigitNet().to(device)
model
```

## 7. Loss и optimizer

Для мультиклассовой классификации использую CrossEntropyLoss. Adam обновляет веса модели во время обучения.

```python
# Loss для задачи классификации на несколько классов
loss_fn = nn.CrossEntropyLoss()
# Optimizer для обновления весов
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

## 8. Функции обучения и оценки

Чтобы код был аккуратнее, отдельно пишу функцию обучения одной эпохи и функцию проверки на validation.

```python
# Функция обучения одной эпохи
def train_epoch(model, loader):
    model.train()
    total_loss = 0
    all_pred = []
    all_true = []
    for xb, yb in loader:
        xb = xb.to(device)
        yb = yb.to(device)
        optimizer.zero_grad()
        out = model(xb)
        loss = loss_fn(out, yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(xb)
        pred = out.argmax(dim=1)
        all_pred.extend(pred.cpu().numpy())
        all_true.extend(yb.cpu().numpy())
    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_true, all_pred)
    return avg_loss, acc
```

```python
# Функция оценки модели
def eval_model(model, loader):
    model.eval()
    total_loss = 0
    all_pred = []
    all_true = []
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)
            out = model(xb)
            loss = loss_fn(out, yb)
            total_loss += loss.item() * len(xb)
            pred = out.argmax(dim=1)
            all_pred.extend(pred.cpu().numpy())
            all_true.extend(yb.cpu().numpy())
    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_true, all_pred)
    return avg_loss, acc, np.array(all_true), np.array(all_pred)
```

## 9. Обучение модели

Обучаю модель несколько эпох и сохраняю loss и accuracy. Эпох немного, чтобы ноутбук запускался быстро на CPU.

```python
# Количество эпох
epochs = 12
# Здесь буду хранить историю обучения
history = []
for epoch in range(1, epochs + 1):
    train_loss, train_acc = train_epoch(model, train_loader)
    val_loss, val_acc, y_true, y_pred = eval_model(model, val_loader)
    history.append({'epoch': epoch, 'train_loss': train_loss, 'val_loss': val_loss, 'train_acc': train_acc, 'val_acc': val_acc})
    if epoch == 1 or epoch % 5 == 0:
        print(f'Epoch {epoch}: train_acc={train_acc:.3f}, val_acc={val_acc:.3f}, val_loss={val_loss:.3f}')
history = pd.DataFrame(history)
history.tail()
```

## 10. Графики обучения

Смотрю, как менялись loss и accuracy. Если train становится лучше, а validation хуже, это признак переобучения.

```python
# График loss
plt.plot(history['epoch'], history['train_loss'], label='train loss')
plt.plot(history['epoch'], history['val_loss'], label='val loss')
plt.title('Loss по эпохам')
plt.xlabel('Эпоха')
plt.ylabel('Loss')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# График accuracy
plt.plot(history['epoch'], history['train_acc'], label='train accuracy')
plt.plot(history['epoch'], history['val_acc'], label='val accuracy')
plt.title('Accuracy по эпохам')
plt.xlabel('Эпоха')
plt.ylabel('Accuracy')
plt.legend()
plt.tight_layout()
plt.show()
```

## 11. Финальная оценка

Смотрю качество модели на validation и отчет по каждому классу.

```python
# Получаю финальные предсказания
val_loss, val_acc, y_true, y_pred = eval_model(model, val_loader)
print(f'Validation loss: {val_loss:.4f}')
print(f'Validation accuracy: {val_acc:.4f}')
print(classification_report(y_true, y_pred))
```

## 12. Матрица ошибок

Матрица ошибок показывает, какие цифры модель путает между собой.

```python
# Строю матрицу ошибок
cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Матрица ошибок')
plt.xlabel('Прогноз')
plt.ylabel('Реальный класс')
plt.tight_layout()
plt.show()
```

## 13. Примеры ошибок

Показываю несколько изображений, где модель ошиблась. Это помогает понять, насколько ошибки выглядят естественными.

```python
# Нахожу ошибки
wrong = np.where(y_true != y_pred)[0]
# Показываю до 10 ошибок
n = min(10, len(wrong))
fig, ax = plt.subplots(2, 5, figsize=(10, 4))
for a in ax.flatten():
    a.axis('off')
for i, a in enumerate(ax.flatten()[:n]):
    idx = wrong[i]
    img = x_val[idx].reshape(8, 8)
    a.imshow(img, cmap='gray')
    a.set_title(f'true {y_true[idx]}, pred {y_pred[idx]}')
    a.axis('off')
plt.tight_layout()
plt.show()
print(f'Ошибок: {len(wrong)} из {len(y_true)}')
```

## Итоги

В этом ноутбуке я обучила простую нейронную сеть на изображениях цифр.

- Данные нужно перевести в тензоры, потому что PyTorch работает с tensor-форматом.
- DataLoader подает данные батчами, что удобно для обучения.
- Для мультиклассовой классификации подходит CrossEntropyLoss.
- Accuracy на train и validation помогает следить за обучением.
- Матрица ошибок показывает, какие классы модель путает.
- Даже простая полносвязная сеть хорошо справляется с маленькими изображениями digits.

Главный вывод: глубокое обучение - это не только архитектура модели. Важны подготовка данных, правильный loss, цикл обучения, контроль качества и анализ ошибок.
