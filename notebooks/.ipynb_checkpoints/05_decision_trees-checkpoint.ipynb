# Деревья решений

**Датасет:** House Prices

В этом ноутбуке я решаю задачу классификации с помощью дерева решений. Цель такая же, как в бинарной классификации: определить, относится ли дом к дорогим.

## План работы

1. Загрузить данные и создать бинарную цель.
2. Подготовить признаки для дерева решений.
3. Обучить базовое дерево.
4. Сравнить качество на train и validation.
5. Проверить влияние глубины дерева.
6. Посмотреть важность признаков.
7. Визуализировать небольшое дерево.
8. Сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для работы с таблицами, графиков, дерева решений и метрик классификации.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.metrics import classification_report, confusion_matrix
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Загрузка данных

Загружаю таблицу с домами. Дорогой дом определяю через цену: если цена выше 75 процентиля, ставлю класс 1, иначе класс 0.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Создаю границу дорогого дома
limit = df['SalePrice'].quantile(0.75)
# Создаю бинарную цель
df['HighPrice'] = (df['SalePrice'] > limit).astype(int)
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонки')
print(f'Граница дорогого дома: {limit:.0f}')
df.head()
```

**Короткий вывод:** теперь задача стала бинарной классификацией: класс 1 означает дорогой дом, класс 0 - остальные дома.

## 3. Баланс классов

Проверяю, сколько дорогих и обычных домов в данных. Для дерева это тоже важно: при дисбалансе модель может чаще выбирать большой класс.

```python
# Считаю количество объектов каждого класса
class_count = df['HighPrice'].value_counts().sort_index()
class_count
```

```python
# Строю график баланса классов
sns.countplot(data=df, x='HighPrice', color='steelblue')
plt.title('Баланс классов')
plt.xlabel('Класс')
plt.ylabel('Количество домов')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** дорогих домов меньше, чем обычных. Поэтому в оценке буду смотреть не только accuracy, но и precision, recall, F1 и ROC-AUC.

## 4. Подготовка признаков

Для дерева решений масштабирование не нужно. Но текстовые признаки все равно нужно превратить в числа, поэтому использую one-hot encoding.

```python
# y - ответ: дорогой дом или нет
y = df['HighPrice']
# x - признаки. Id, SalePrice и HighPrice убираю
x = df.drop(columns=['Id', 'SalePrice', 'HighPrice'])
print(f'x: {x.shape}')
print(f'y: {y.shape}')
```

```python
# Разделяю колонки на числовые и текстовые
num_cols = x.select_dtypes(include='number').columns
cat_cols = x.select_dtypes(include='object').columns
print(f'Числовых колонок: {len(num_cols)}')
print(f'Текстовых колонок: {len(cat_cols)}')
```

```python
# Числовые пропуски заполняю медианой
x[num_cols] = x[num_cols].fillna(x[num_cols].median())
```

```python
# Текстовые пропуски заполняю словом None
x[cat_cols] = x[cat_cols].fillna('None')
```

```python
# Кодирую текстовые признаки
x = pd.get_dummies(x, drop_first=True)
print(f'Размер x после кодирования: {x.shape}')
print(f'Осталось пропусков: {x.isna().sum().sum()}')
```

## 5. Разделение на train и validation

Делю данные на обучение и проверку. Stratify сохраняет похожую долю дорогих домов в обеих частях.

```python
# 80 процентов данных идут на обучение, 20 процентов - на проверку
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
print('Доля класса 1 в train:', y_train.mean().round(3))
print('Доля класса 1 в val:  ', y_val.mean().round(3))
```

## 6. Базовое дерево решений

Сначала обучаю дерево с ограниченной глубиной. Это помогает снизить переобучение: дерево не сможет запомнить все мелкие детали train.

```python
# Создаю дерево с ограниченной глубиной
tree = DecisionTreeClassifier(max_depth=4, min_samples_leaf=10, random_state=42)
# Обучаю дерево на train
tree.fit(x_train, y_train)
```

```python
# Делаю прогнозы классов и вероятностей
pred = tree.predict(x_val)
prob = tree.predict_proba(x_val)[:, 1]
# Считаю основные метрики
print('Accuracy:', round(accuracy_score(y_val, pred), 4))
print('Precision:', round(precision_score(y_val, pred), 4))
print('Recall:', round(recall_score(y_val, pred), 4))
print('F1:', round(f1_score(y_val, pred), 4))
print('ROC-AUC:', round(roc_auc_score(y_val, prob), 4))
```

```python
# Вывожу подробный отчет по классам
print(classification_report(y_val, pred))
```

**Короткий вывод:** дерево дает понятную базовую модель. Но для деревьев особенно важно контролировать глубину, иначе модель легко переобучается.

## 7. Проверка переобучения

Сравниваю качество на train и validation при разной глубине дерева. Если train растет, а validation падает, значит дерево начинает запоминать обучающую выборку.

```python
# Проверяю разные значения max_depth
rows = []
for depth in range(1, 16):
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    model.fit(x_train, y_train)
    train_pred = model.predict(x_train)
    val_pred = model.predict(x_val)
    rows.append({'max_depth': depth, 'train_f1': f1_score(y_train, train_pred), 'val_f1': f1_score(y_val, val_pred)})
# Собираю результаты в таблицу
depth_df = pd.DataFrame(rows)
depth_df.head()
```

```python
# Строю график зависимости F1 от глубины дерева
plt.plot(depth_df['max_depth'], depth_df['train_f1'], marker='o', label='train F1')
plt.plot(depth_df['max_depth'], depth_df['val_f1'], marker='o', label='validation F1')
plt.title('Влияние глубины дерева на качество')
plt.xlabel('max_depth')
plt.ylabel('F1')
plt.legend()
plt.tight_layout()
plt.show()
```

**Короткий вывод:** при большой глубине качество на train обычно растет, но validation может перестать улучшаться. Это типичный признак переобучения дерева.

## 8. Лучшая глубина по validation

Выбираю глубину, где F1 на validation получился максимальным, и обучаю дерево с этой глубиной.

```python
# Нахожу лучшую глубину по validation F1
best_depth = int(depth_df.sort_values('val_f1', ascending=False).iloc[0]['max_depth'])
print(f'Лучшая глубина: {best_depth}')
# Обучаю дерево с лучшей глубиной
best_tree = DecisionTreeClassifier(max_depth=best_depth, min_samples_leaf=10, random_state=42)
best_tree.fit(x_train, y_train)
# Считаю метрики лучшего дерева
pred = best_tree.predict(x_val)
prob = best_tree.predict_proba(x_val)[:, 1]
print('Accuracy:', round(accuracy_score(y_val, pred), 4))
print('Precision:', round(precision_score(y_val, pred), 4))
print('Recall:', round(recall_score(y_val, pred), 4))
print('F1:', round(f1_score(y_val, pred), 4))
print('ROC-AUC:', round(roc_auc_score(y_val, prob), 4))
```

## 9. Матрица ошибок

Матрица ошибок показывает, сколько дорогих и обычных домов дерево определило правильно и неправильно.

```python
# Строю матрицу ошибок
cm = confusion_matrix(y_val, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Матрица ошибок')
plt.xlabel('Прогноз')
plt.ylabel('Реальный класс')
plt.tight_layout()
plt.show()
```

## 10. Важность признаков

Дерево умеет показывать, какие признаки чаще всего помогали делать разделения. Это удобно для интерпретации модели.

```python
# Собираю важности признаков в таблицу
imp = pd.DataFrame({'feature': x_train.columns, 'importance': best_tree.feature_importances_})
imp = imp.sort_values('importance', ascending=False).head(15)
imp
```

```python
# Строю график топ-15 важных признаков
sns.barplot(data=imp, x='importance', y='feature', color='steelblue')
plt.title('Топ-15 важных признаков')
plt.xlabel('Важность')
plt.ylabel('Признак')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** важность признаков помогает объяснить модель. Обычно наверху оказываются качество дома, площадь, гараж и другие признаки, которые логично связаны с дорогими домами.

## 11. Визуализация дерева

Показываю небольшое дерево с глубиной 3. Его проще читать: видно, по каким условиям модель разделяет дома.

```python
# Обучаю маленькое дерево специально для визуализации
small_tree = DecisionTreeClassifier(max_depth=3, min_samples_leaf=20, random_state=42)
small_tree.fit(x_train, y_train)
# Рисую дерево
plt.figure(figsize=(18, 8))
plot_tree(small_tree, feature_names=x_train.columns, class_names=['обычный', 'дорогой'], filled=True, rounded=True, fontsize=8)
plt.title('Небольшое дерево решений')
plt.show()
```

```python
# Показываю текстовую версию дерева
text_tree = export_text(small_tree, feature_names=list(x_train.columns), max_depth=3)
print(text_tree)
```

## Итоги

В этом ноутбуке я построила дерево решений для классификации дорогих домов.

- Дерево решений не требует масштабирования признаков.
- Категориальные признаки все равно нужно закодировать в числа.
- Глубину дерева важно ограничивать, потому что глубокое дерево легко переобучается.

- Важность признаков делает модель более понятной.
- Визуализация маленького дерева помогает объяснить логику решений.

Главный вывод: дерево решений удобно тем, что его можно интерпретировать, но без ограничения глубины оно быстро начинает запоминать train и хуже обобщать на новые данные.
