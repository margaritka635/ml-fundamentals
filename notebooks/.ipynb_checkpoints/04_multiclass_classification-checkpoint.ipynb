# Мультиклассовая классификация

**Датасет:** Digits из sklearn

В этом ноутбуке я решаю задачу мультиклассовой классификации: нужно определить цифру на маленьком изображении. Классов здесь 10: цифры от 0 до 9.

## План работы

1. Загрузить датасет с изображениями цифр.
2. Посмотреть размер данных и примеры картинок.
3. Проверить баланс классов.
4. Разделить данные на train и validation.
5. Масштабировать признаки.
6. Обучить LogisticRegression, KNN и SVM.
7. Сравнить модели по accuracy и macro F1.
8. Посмотреть отчет по лучшей модели и сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для работы с таблицами, графиков, загрузки данных, обучения моделей и оценки качества.

```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_digits

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

from sklearn.preprocessing import StandardScaler

%matplotlib inline

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Загрузка данных

Загружаю встроенный датасет digits. Каждая картинка - это цифра размером 8 на 8 пикселей. Для модели картинка представлена как 64 числовых признака.

```python
# Загружаю данные
data = load_digits()

# x - признаки, y - правильные классы цифр
x = data.data
y = data.target

print(f'x: {x.shape}')
print(f'y: {y.shape}')
print(f'Классы: {data.target_names}')
```

**Короткий вывод:** в данных 1797 изображений и 64 признака. Нужно предсказать один из 10 классов.

## 3. Примеры изображений

Перед обучением полезно посмотреть на сами картинки. Так проще понять, что именно классифицирует модель.

```python
# Показываю первые 10 изображений из датасета
fig, ax = plt.subplots(2, 5, figsize=(10, 4))

for i, a in enumerate(ax.flatten()):
    a.imshow(data.images[i], cmap='gray')
    a.set_title(f'Класс: {y[i]}')
    a.axis('off')

plt.tight_layout()
plt.show()
```

**Короткий вывод:** изображения маленькие и немного грубые, но цифры хорошо различимы. Это подходит для базового сравнения моделей.

## 4. Баланс классов

Проверяю, сколько примеров есть для каждой цифры. Если классы распределены примерно ровно, accuracy будет более честной метрикой.

```python
# Считаю количество объектов каждого класса
class_count = pd.Series(y).value_counts().sort_index()
class_count
```

```python
# Строю график распределения классов
sns.barplot(x=class_count.index, y=class_count.values, color='steelblue')
plt.title('Количество объектов по классам')
plt.xlabel('Класс')
plt.ylabel('Количество')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** классы распределены почти равномерно. Значит, accuracy можно использовать, но macro F1 тоже оставлю, чтобы все классы учитывались одинаково.

## 5. Разделение на train и validation

Делю данные на обучающую и проверочную части. Stratify сохраняет похожие доли классов в обеих частях.

```python
# 80 процентов данных идут на обучение, 20 процентов - на проверку
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)

print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
```

## 6. Масштабирование признаков

Для LogisticRegression и SVM масштабирование важно, потому что признаки должны быть в похожем диапазоне. KNN тоже часто работает лучше после масштабирования.

```python
# Создаю scaler и обучаю его только на train
scaler = StandardScaler()

# Масштабирую train и validation одинаковым способом
x_train_sc = scaler.fit_transform(x_train)
x_val_sc = scaler.transform(x_val)

print(f'Размер train после scaling: {x_train_sc.shape}')
print(f'Размер val после scaling:   {x_val_sc.shape}')
```

## 7. Обучение моделей

Сравниваю три модели: логистическую регрессию, KNN и SVM. Все они умеют решать задачу с несколькими классами.

```python
# Создаю модели для сравнения
models = {
    'LogisticRegression': LogisticRegression(max_iter=2000),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'SVM': SVC(kernel='rbf', probability=True)
}

models
```

```python
# В эту таблицу буду складывать качество моделей
res = []

# Обучаю каждую модель и считаю метрики на validation
for name, model in models.items():
    model.fit(x_train_sc, y_train)
    pred = model.predict(x_val_sc)
    acc = accuracy_score(y_val, pred)
    f1 = f1_score(y_val, pred, average='macro')
    res.append({'model': name, 'accuracy': acc, 'macro_f1': f1})

# Собираю результаты в таблицу и сортирую по macro F1
res = pd.DataFrame(res).sort_values('macro_f1', ascending=False)
res
```

**Короткий вывод:** accuracy показывает общую долю правильных ответов. Macro F1 считает качество по каждому классу и потом усредняет, поэтому он хорошо подходит для мультиклассовой задачи.

## 8. Сравнение моделей на графике

Строю простой график, чтобы было легче сравнить качество моделей.

```python
# Строю график macro F1
sns.barplot(data=res, x='macro_f1', y='model', color='steelblue')
plt.title('Сравнение моделей по macro F1')
plt.xlabel('Macro F1')
plt.ylabel('Модель')
plt.tight_layout()
plt.show()
```

## 9. Подробная оценка лучшей модели

Беру лучшую модель по macro F1 и смотрю отчет по каждому классу. Так видно, какие цифры модель распознает лучше или хуже.

```python
# Выбираю лучшую модель
best_name = res.iloc[0]['model']
best_model = models[best_name]

# Делаю прогноз лучшей модели
pred = best_model.predict(x_val_sc)

print(f'Лучшая модель: {best_name}')
print(classification_report(y_val, pred))
```

## 10. Матрица ошибок

Матрица ошибок показывает, какие цифры модель путает между собой.

```python
# Строю матрицу ошибок для лучшей модели
cm = confusion_matrix(y_val, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title(f'Матрица ошибок: {best_name}')
plt.xlabel('Прогноз')
plt.ylabel('Реальный класс')
plt.tight_layout()
plt.show()
```

## 11. Примеры ошибок

Посмотрю несколько картинок, где лучшая модель ошиблась. Это помогает понять, насколько ошибки выглядят естественными.

```python
# Нахожу индексы ошибок в validation
wrong = np.where(pred != y_val)[0]

# Показываю до 10 ошибочных примеров
n = min(10, len(wrong))
fig, ax = plt.subplots(2, 5, figsize=(10, 4))

for a in ax.flatten():
    a.axis('off')

for i, a in enumerate(ax.flatten()[:n]):
    idx = wrong[i]
    img = x_val[idx].reshape(8, 8)
    a.imshow(img, cmap='gray')
    a.set_title(f'true {y_val[idx]}, pred {pred[idx]}')
    a.axis('off')

plt.tight_layout()
plt.show()
print(f'Количество ошибок: {len(wrong)} из {len(y_val)}')
```

## Итоги

В этом ноутбуке я построила решение для мультиклассовой классификации цифр.

- В датасете 10 классов: цифры от 0 до 9.
- Классы распределены почти равномерно, поэтому accuracy можно использовать без сильного искажения.
- Признаки масштабированы, потому что это важно для LogisticRegression, KNN и SVM.
- Сравнение моделей сделано по accuracy и macro F1.
- Матрица ошибок помогает увидеть, какие цифры модель путает.

Главный вывод: для мультиклассовой классификации важно смотреть не только общую accuracy, но и качество по каждому классу. Даже высокая accuracy может скрывать ошибки на отдельных классах.
