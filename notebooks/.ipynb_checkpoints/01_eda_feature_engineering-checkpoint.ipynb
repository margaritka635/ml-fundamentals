# EDA и Feature Engineering

**Датасет:** House Prices — Advanced Regression Techniques (Kaggle)

В этом ноутбуке я разбираю данные о продаже домов: сначала смотрю на качество данных, потом проверяю связь признаков с ценой, создаю новые признаки и готовлю таблицу для будущей модели.

## План работы

1. Загрузить train и test, проверить размеры таблиц.
2. Посмотреть на целевую переменную SalePrice.
3. Найти пропуски и решить, как их заполнить.
4. Проверить числовые и категориальные признаки.
5. Создать новые признаки, которые лучше описывают дом.
6. Закодировать категориальные признаки без искусственного порядка.
7. Разделить данные обратно на train и test.

## 1. Импорт библиотек

Здесь подключаю библиотеки для работы с таблицами, расчетов и графиков. Настройки нужны только для того, чтобы вывод был удобным.

```python
# pandas нужен для работы с таблицами.
import pandas as pd

# numpy нужен для математических расчетов, например для логарифма цены.
import numpy as np

# matplotlib и seaborn нужны для графиков.
import matplotlib.pyplot as plt
import seaborn as sns

# Показываю графики прямо в ноутбуке.
%matplotlib inline

# Показываю больше колонок, чтобы удобнее смотреть широкую таблицу.
pd.set_option('display.max_columns', 120)

# Задаю единый стиль графиков.
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Загрузка данных

Загружаю обучающую и тестовую таблицы. В train есть цена дома, а в test цены нет, потому что для test ее потом нужно предсказывать.

```python
# Читаю исходные csv-файлы из папки data.
train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')

# Запоминаю размер train, чтобы потом разделить общую таблицу обратно.
n_train = train.shape[0]

# Проверяю размеры таблиц.
print(f'Train: {train.shape[0]} строк, {train.shape[1]} колонок')
print(f'Test:  {test.shape[0]} строк, {test.shape[1]} колонок')

# Смотрю первые строки таблицы.
train.head()
```

**Короткий вывод:** данные загрузились корректно. В train на одну колонку больше, потому что там есть цена продажи дома.

## 3. Общая структура данных

Проверяю, сколько признаков числовые, сколько категориальные, и есть ли повторы по Id. Это базовая проверка перед анализом.

```python
# Считаю количество числовых и текстовых признаков.
num_count = train.select_dtypes(include='number').shape[1]
cat_count = train.select_dtypes(include='object').shape[1]

# Проверяю, повторяется ли Id.
id_dup = train['Id'].duplicated().sum()

print(f'Числовые признаки:       {num_count}')
print(f'Категориальные признаки: {cat_count}')
print(f'Дубликаты по Id:         {id_dup}')
```

**Короткий вывод:** в данных есть и числовые, и категориальные признаки. Поэтому дальше их лучше анализировать отдельно.

## 4. Анализ цены

Цена дома — это целевая переменная. Сначала смотрю основные статистики и распределение цены. Еще сравниваю исходную цену с логарифмом цены, потому что цены часто имеют длинный правый хвост.

```python
# Вывожу основные статистики по цене.
print(train['SalePrice'].describe())

# Строю два графика рядом: исходная цена и логарифм цены.
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].hist(train['SalePrice'], bins=50, color='steelblue', edgecolor='white')
ax[0].set_title('SalePrice: исходное распределение')
ax[0].set_xlabel('Цена продажи')
ax[0].set_ylabel('Количество домов')

ax[1].hist(np.log1p(train['SalePrice']), bins=50, color='coral', edgecolor='white')
ax[1].set_title('log1p(SalePrice): после логарифма')
ax[1].set_xlabel('Логарифм цены')
ax[1].set_ylabel('Количество домов')

plt.tight_layout()
plt.show()

# Асимметрия показывает, насколько распределение скошено.
print(f"Асимметрия исходной цены: {train['SalePrice'].skew():.2f}")
print(f"Асимметрия после логарифма: {np.log1p(train['SalePrice']).skew():.2f}")
```

**Короткий вывод:** исходная цена скошена вправо: дорогих домов мало, но они сильно растягивают распределение. После логарифма распределение становится ровнее, поэтому для будущей регрессии это хороший вариант.

## 5. Пропуски

Смотрю, где в данных есть пропуски. В этом датасете часть пропусков означает не ошибку, а отсутствие объекта: например, нет бассейна, гаража или подвала.

```python
# Считаю количество пропусков в каждой колонке train.
miss = train.isna().sum()
miss = miss[miss > 0].sort_values(ascending=False)

# Перевожу количество пропусков в проценты.
miss_pct = (miss / len(train) * 100).round(1)

# Собираю таблицу с количеством и процентом пропусков.
miss_df = pd.DataFrame({
    'missing_count': miss,
    'missing_percent': miss_pct
})

# Показываю признаки с самым большим числом пропусков.
miss_df.head(20)
```

```python
# Строю график для топ-20 признаков с пропусками.
miss_df.head(20)['missing_percent'].sort_values().plot(kind='barh', color='salmon', figsize=(10, 6))

plt.title('Топ-20 признаков по доле пропусков')
plt.xlabel('Доля пропусков, %')
plt.ylabel('Признак')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** больше всего пропусков у редких объектов: бассейн, забор, аллея, дополнительные постройки. Здесь пропуск часто означает, что объекта просто нет.

## 6. Числовые признаки и цена

Для числовых признаков считаю корреляцию с ценой. Это не доказывает причину, но помогает быстро увидеть, какие признаки сильнее всего связаны с ценой дома.

```python
# Беру числовые признаки и убираю Id и SalePrice.
num_cols = train.select_dtypes(include='number').columns.tolist()
num_cols = [col for col in num_cols if col not in ['Id', 'SalePrice']]

# Считаю корреляцию каждого числового признака с ценой.
corr = train[num_cols + ['SalePrice']].corr()['SalePrice'].drop('SalePrice')

# Беру 15 признаков с самой сильной связью по модулю.
top_corr = corr.abs().sort_values(ascending=False).head(15)

# Показываю знак корреляции на графике.
corr[top_corr.index].sort_values().plot(kind='barh', color='steelblue', figsize=(10, 7))

plt.title('Топ-15 числовых признаков по корреляции с ценой')
plt.xlabel('Корреляция')
plt.ylabel('Признак')
plt.tight_layout()
plt.show()

# Вывожу точные значения корреляции.
corr[top_corr.index].sort_values(ascending=False).round(3)
```

**Короткий вывод:** сильнее всего с ценой связаны качество дома, жилая площадь, гараж и площадь подвала. Это выглядит логично: более качественные и большие дома обычно дороже.

## 7. Связи между сильными признаками

Некоторые признаки похожи друг на друга. Например, количество машин в гараже и площадь гаража описывают почти одну и ту же часть дома. Тепловая карта помогает увидеть такие связи.

```python
# Беру признаки, которые сильнее всего связаны с ценой.
top15 = top_corr.index.tolist()

# Строю тепловую карту корреляций.
plt.figure(figsize=(14, 10))
sns.heatmap(train[top15 + ['SalePrice']].corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)

plt.title('Корреляции между сильными числовыми признаками')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** часть признаков несет похожую информацию. Для деревьев и бустинга это обычно не страшно, а для линейных моделей позже может пригодиться регуляризация.

## 8. Категориальные признаки

Для категориальных признаков сравниваю цену внутри разных групп. Так видно, какие категории реально разделяют дешевые и дорогие дома.

```python
# Выбираю несколько понятных категориальных признаков.
cat_show = ['Neighborhood', 'BldgType', 'HouseStyle', 'SaleCondition', 'KitchenQual', 'ExterQual']

# Для каждого признака строю boxplot.
fig, ax = plt.subplots(2, 3, figsize=(16, 10))

for a, col in zip(ax.flatten(), cat_show):
    # Сортирую категории по медианной цене, чтобы график было проще читать.
    order = train.groupby(col)['SalePrice'].median().sort_values(ascending=False).index
    
    sns.boxplot(data=train, x=col, y='SalePrice', order=order, ax=a, hue=col, palette='Blues_r', legend=False)
    
    a.set_title(col)
    a.set_xlabel('')
    a.set_ylabel('SalePrice')
    a.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

**Короткий вывод:** район, качество кухни и качество внешней отделки заметно связаны с ценой. Значит, текстовые признаки тоже полезны для модели.

## 9. Общая таблица для обработки

Перед обработкой объединяю train и test без цены. Так пропуски и категории будут обработаны одинаково в обеих таблицах.

```python
# Отделяю целевую переменную, чтобы не смешивать ее с признаками.
y = train['SalePrice']
y_log = np.log1p(y)

# Объединяю признаки train и test в одну таблицу.
df = pd.concat([train.drop(columns=['SalePrice']), test], axis=0, ignore_index=True)

print(f'Общая таблица: {df.shape[0]} строк, {df.shape[1]} колонок')
```

## 10. Заполнение пропусков

Заполняю пропуски не одним способом для всех колонок, а по смыслу. Если объекта нет, ставлю None или 0. Если значение просто пропущено, использую медиану или самое частое значение.

```python
# В этих текстовых колонках пропуск означает, что объекта нет.
none_cols = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
    'MasVnrType'
]

none_cols
```

```python
# Заполняю такие текстовые пропуски словом None.
for col in none_cols:
    df[col] = df[col].fillna('None')
```

```python
# В этих числовых колонках пропуск означает отсутствие связанного объекта.
zero_cols = [
    'GarageYrBlt', 'GarageArea', 'GarageCars',
    'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF',
    'BsmtFullBath', 'BsmtHalfBath', 'MasVnrArea'
]

zero_cols
```

```python
# Заполняю такие числовые пропуски нулем.
for col in zero_cols:
    df[col] = df[col].fillna(0)
```

```python
# LotFrontage зависит от района, поэтому беру медиану внутри каждого района.
df['LotFrontage'] = df.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
# Остальные числовые пропуски заполняю медианой по колонке.
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
# Остальные текстовые пропуски заполняю самым частым значением.
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Проверяю, что пропусков больше нет.
print(f'Осталось пропусков: {df.isna().sum().sum()}')
```

**Короткий вывод:** пропуски заполнены по смыслу. Так мы меньше искажаем данные и не теряем много строк.

## 11. Создание новых признаков

Создаю простые признаки, которые легко объяснить: общая площадь, количество ванных комнат, площадь крыльца, возраст дома и наличие важных объектов.

```python
# Общая площадь дома: подвал + первый этаж + второй этаж.
df['TotalSF'] = df['TotalBsmtSF'] + df['1stFlrSF'] + df['2ndFlrSF']
```

```python
# Общее число ванных комнат. Половинные ванны считаю как 0.5.
df['TotalBath'] = (
    df['FullBath'] +
    0.5 * df['HalfBath'] +
    df['BsmtFullBath'] +
    0.5 * df['BsmtHalfBath']
)
```

```python
# Общая площадь крыльца и веранд.
df['TotalPorchSF'] = (
    df['OpenPorchSF'] +
    df['EnclosedPorch'] +
    df['3SsnPorch'] +
    df['ScreenPorch']
)
```

```python
# Возраст дома на момент продажи.
df['HouseAge'] = df['YrSold'] - df['YearBuilt']

# Сколько лет прошло после последнего ремонта.
df['RemodAge'] = df['YrSold'] - df['YearRemodAdd']
```

```python
# Признак показывает, был ли ремонт.
df['IsRemodeled'] = (df['YearBuilt'] != df['YearRemodAdd']).astype(int)
```

```python
# Признаки наличия гаража, бассейна и камина.
df['HasGarage'] = (df['GarageArea'] > 0).astype(int)
df['HasPool'] = (df['PoolArea'] > 0).astype(int)
df['HasFireplace'] = (df['Fireplaces'] > 0).astype(int)

print(f'После создания новых признаков: {df.shape[1]} колонок')
```

**Короткий вывод:** новые признаки собирают разрозненную информацию в более понятные показатели. Например, модели проще использовать общую площадь, чем отдельно складывать этажи и подвал.

## 12. Кодирование категориальных признаков

Для категориальных признаков использую one-hot encoding. Так я не задаю ложный порядок там, где его нет: например, один район не должен автоматически быть больше другого района.

```python
# Превращаю текстовые признаки в набор бинарных колонок.
df_enc = pd.get_dummies(df, drop_first=False)
```

```python
# Разделяю общую таблицу обратно на train и test.
X_train = df_enc.iloc[:n_train].copy()
X_test = df_enc.iloc[n_train:].copy()

print(f'X_train: {X_train.shape}')
print(f'X_test:  {X_test.shape}')
print(f'y:       {y.shape}')
print(f'y_log:   {y_log.shape}')
```

**Короткий вывод:** после кодирования все признаки стали числовыми. Теперь X_train, X_test, y и y_log можно использовать дальше для обучения моделей.

## Итоговые выводы

- В данных 1460 объектов для обучения и 1459 объектов для теста.
- Цена продажи сильно скошена вправо, поэтому для регрессии лучше использовать логарифм цены.
- Самые сильные числовые признаки: качество дома, жилая площадь, гараж и площадь подвала.
- Пропуски часто означают отсутствие объекта, поэтому часть колонок логично заполнять None и 0.
- Категориальные признаки важны: район и качество отделки хорошо разделяют дома по цене.
- Новые признаки вроде общей площади, количества ванных комнат и возраста дома делают данные понятнее.
- Для категорий выбран one-hot encoding, потому что он не создает ложный порядок между категориями.

**Главная мысль:** данные стали чище и понятнее: пропуски обработаны по смыслу, важные закономерности проверены, а признаки подготовлены для будущей модели.
