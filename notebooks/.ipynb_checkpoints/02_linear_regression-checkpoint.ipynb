# Линейная регрессия

**Датасет:** House Prices

В этом ноутбуке я обучаю несколько линейных моделей для прогноза цены дома. Сначала готовлю данные, потом сравниваю обычную линейную регрессию с Ridge, Lasso и ElasticNet.

## План работы

1. Загрузить данные и быстро проверить таблицу.
2. Подготовить признаки и целевую переменную.
3. Заполнить пропуски и закодировать категории.
4. Разделить данные на train и validation.
5. Обучить LinearRegression, Ridge, Lasso и ElasticNet.
6. Сравнить модели по RMSE и R2.
7. Сделать короткие выводы.

## 1. Импорт библиотек

Подключаю библиотеки для работы с таблицами, графиков, разделения данных, масштабирования признаков и обучения линейных моделей.

```python
# pandas нужен для работы с таблицами.
import pandas as pd

# numpy нужен для расчетов.
import numpy as np

# matplotlib и seaborn нужны для графиков.
import matplotlib.pyplot as plt
import seaborn as sns

# Линейные модели, которые будем сравнивать.
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet

# train_test_split нужен для разделения данных на обучение и проверку.
from sklearn.model_selection import train_test_split

# Метрики качества для задачи регрессии.
from sklearn.metrics import mean_squared_error, r2_score

# StandardScaler нужен для масштабирования признаков перед регуляризацией.
from sklearn.preprocessing import StandardScaler

# Показываю графики прямо в ноутбуке.
%matplotlib inline

# Настраиваю внешний вид таблиц и графиков.
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Загрузка данных

Загружаю обучающую таблицу. В ней есть признаки дома и цена продажи, которую нужно предсказывать.

```python
# Читаю данные из общей папки data.
df = pd.read_csv('../data/train.csv')

# Проверяю размер таблицы.
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонок')

# Смотрю первые строки, чтобы быстро проверить формат данных.
df.head()
```

**Короткий вывод:** данные загрузились корректно. В таблице есть числовые и категориальные признаки, поэтому перед моделью их нужно подготовить.

## 3. Быстрая проверка данных

Смотрю базовую информацию: сколько признаков разных типов, есть ли дубликаты по Id и как выглядит распределение цены.

```python
# Считаю числовые и категориальные признаки.
num_count = df.select_dtypes(include='number').shape[1]
cat_count = df.select_dtypes(include='object').shape[1]

# Проверяю дубликаты по Id.
id_dup = df['Id'].duplicated().sum()

print(f'Числовые признаки:       {num_count}')
print(f'Категориальные признаки: {cat_count}')
print(f'Дубликаты по Id:         {id_dup}')
```

```python
# Смотрю основные статистики по цене.
print(df['SalePrice'].describe())

# Строю график цены.
plt.hist(df['SalePrice'], bins=50, color='steelblue', edgecolor='white')
plt.title('Распределение цены продажи')
plt.xlabel('Цена')
plt.ylabel('Количество домов')
plt.tight_layout()
plt.show()

# Проверяю асимметрию цены.
print(f"Асимметрия цены: {df['SalePrice'].skew():.2f}")
```

**Короткий вывод:** цена сильно скошена вправо. Для линейной регрессии это неудобно, поэтому дальше я буду предсказывать логарифм цены, а не исходную цену.

## 4. Подготовка цели и признаков

Отделяю цену от признаков. Для цели беру логарифм цены, потому что так распределение становится ровнее и модели проще учиться.

```python
# y — целевая переменная, то есть цена дома после логарифма.
y = np.log1p(df['SalePrice'])

# x — признаки. Id и SalePrice убираю: Id не помогает модели, а SalePrice является ответом.
x = df.drop(columns=['Id', 'SalePrice'])

print(f'x: {x.shape}')
print(f'y: {y.shape}')
```

## 5. Заполнение пропусков

Заполняю пропуски до кодирования категорий. Здесь я использую только признаки, а цену не трогаю. Для числовых признаков беру медиану, для текстовых — отдельное значение None.

```python
# Отдельно сохраняю списки числовых и категориальных колонок.
num_cols = x.select_dtypes(include='number').columns
cat_cols = x.select_dtypes(include='object').columns

print(f'Числовых колонок: {len(num_cols)}')
print(f'Категориальных колонок: {len(cat_cols)}')
```

```python
# Числовые пропуски заполняю медианой.
x[num_cols] = x[num_cols].fillna(x[num_cols].median())
```

```python
# Текстовые пропуски заполняю словом None.
x[cat_cols] = x[cat_cols].fillna('None')
```

```python
# Проверяю, что пропусков больше нет.
print(f'Осталось пропусков: {x.isna().sum().sum()}')
```

**Короткий вывод:** пропуски заполнены простым и понятным способом. Теперь данные можно кодировать и отдавать в модель.

## 6. Кодирование категориальных признаков

Линейная регрессия работает только с числами. Поэтому текстовые признаки превращаю в набор бинарных колонок с помощью one-hot encoding. Одну категорию из каждого признака убираю, чтобы снизить лишнюю зависимость между колонками.

```python
# Превращаю текстовые признаки в числовые бинарные колонки.
x = pd.get_dummies(x, drop_first=True)

print(f'Размер x после кодирования: {x.shape}')
```

## 7. Разделение на train и validation

Теперь разделяю данные на обучающую и проверочную части. Validation нужен, чтобы оценить качество модели на данных, которые она не видела при обучении.

```python
# 80 процентов данных идут на обучение, 20 процентов — на проверку.
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42)

print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
```

## 8. Масштабирование признаков

Ridge, Lasso и ElasticNet чувствительны к масштабу признаков. Поэтому для них делаю отдельные масштабированные версии train и validation. Обычную LinearRegression оставлю на исходных закодированных признаках.

```python
# Создаю scaler и обучаю его только на x_train.
scaler = StandardScaler()

# Масштабирую train и validation одинаковым способом.
x_train_sc = scaler.fit_transform(x_train)
x_val_sc = scaler.transform(x_val)

print(f'Размер train после scaling: {x_train_sc.shape}')
print(f'Размер val после scaling:   {x_val_sc.shape}')
```

## 9. Обучение моделей

Обучаю четыре модели. LinearRegression будет базовой моделью, а Ridge, Lasso и ElasticNet добавляют регуляризацию.

```python
# Создаю словарь моделей, чтобы обучать их в одном цикле.
models = {
    'LinearRegression': LinearRegression(),
    'Ridge': Ridge(alpha=10),
    'Lasso': Lasso(alpha=0.001, max_iter=10000),
    'ElasticNet': ElasticNet(alpha=0.001, l1_ratio=0.5, max_iter=10000)
}

models
```

```python
# В эту таблицу буду складывать качество моделей.
res = []

# Обучаю каждую модель и считаю качество на validation.
for name, model in models.items():
    if name == 'LinearRegression':
        model.fit(x_train, y_train)
        pred = model.predict(x_val)
    else:
        model.fit(x_train_sc, y_train)
        pred = model.predict(x_val_sc)
    
    rmse = np.sqrt(mean_squared_error(y_val, pred))
    r2 = r2_score(y_val, pred)
    res.append({'model': name, 'rmse_log': rmse, 'r2': r2})

# Собираю результаты в таблицу и сортирую по RMSE.
res = pd.DataFrame(res).sort_values('rmse_log')
res
```

**Короткий вывод:** RMSE считается на логарифме цены. Чем меньше RMSE, тем лучше. R2 показывает, какую долю разброса цены объясняет модель: чем ближе к 1, тем лучше.

## 10. Сравнение моделей на графике

Строю простой график, чтобы быстрее сравнить модели между собой.

```python
# Строю график RMSE для всех моделей.
sns.barplot(data=res, x='rmse_log', y='model', color='steelblue')
plt.title('Сравнение моделей по RMSE на логарифме цены')
plt.xlabel('RMSE log')
plt.ylabel('Модель')
plt.tight_layout()
plt.show()
```

## 11. Проверка лучшей модели

Беру модель с минимальным RMSE и смотрю, как ее прогнозы соотносятся с реальными значениями на validation.

```python
# Выбираю название лучшей модели из таблицы результатов.
best_name = res.iloc[0]['model']
best_model = models[best_name]

# Делаю прогноз лучшей модели.
if best_name == 'LinearRegression':
    pred_log = best_model.predict(x_val)
else:
    pred_log = best_model.predict(x_val_sc)

# Возвращаю логарифмы обратно в обычные цены.
pred_price = np.expm1(pred_log)
true_price = np.expm1(y_val)

print(f'Лучшая модель: {best_name}')
print(f'RMSE на логарифме цены: {res.iloc[0]["rmse_log"]:.4f}')
print(f'R2: {res.iloc[0]["r2"]:.4f}')
```

```python
# Строю график реальных цен против прогнозов.
plt.scatter(true_price, pred_price, alpha=0.6, color='steelblue')
plt.plot([true_price.min(), true_price.max()], [true_price.min(), true_price.max()], color='red')
plt.title('Реальная цена и прогноз лучшей модели')
plt.xlabel('Реальная цена')
plt.ylabel('Прогноз')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** если точки лежат близко к красной линии, модель хорошо попадает в цену. Сильные отклонения чаще появляются на дорогих домах, потому что таких объектов меньше.

## Итоги

В этом ноутбуке я построила базовое решение для задачи регрессии цены дома.

- Целевую переменную лучше брать в логарифме, потому что исходная цена сильно скошена вправо.
- Пропуски заполнены простым способом: медиана для чисел и None для текстовых признаков.
- Категориальные признаки закодированы через one-hot encoding.
- Для Ridge, Lasso и ElasticNet признаки масштабированы, потому что регуляризация чувствительна к масштабу.
- Лучшую модель выбираю по минимальному RMSE на логарифме цены.

Главный вывод: регуляризация помогает линейным моделям на широких таблицах, где после кодирования категорий появляется много признаков.
