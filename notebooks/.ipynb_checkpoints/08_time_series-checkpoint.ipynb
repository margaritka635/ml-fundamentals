# Временные ряды

**Датасет:** House Prices

В этом ноутбуке я строю временной ряд из продаж домов по месяцам. Цель - прогнозировать среднюю цену продажи на следующие месяцы и сравнить несколько подходов.

## План работы

1. Загрузить данные и собрать месячный временной ряд.
2. Посмотреть динамику средней цены.
3. Сделать STL-декомпозицию.
4. Проверить стационарность через ADF-тест.
5. Обучить SARIMAX.
6. Показать Prophet как опциональный подход.
7. Сделать ML-подход с лаговыми признаками.
8. Сравнить модели по MAE и RMSE.
9. Сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, анализа временных рядов, моделей и метрик.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import STL
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.stattools import adfuller
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import importlib.util
has_prophet = importlib.util.find_spec('prophet') is not None
run_prophet = False
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
print('Prophet доступен:', has_prophet)
print('Prophet будет запущен:', run_prophet)
```

## 2. Загрузка данных

В исходной таблице есть год продажи, месяц продажи и цена. Из них можно собрать временной ряд средней цены по месяцам.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Создаю дату из года и месяца продажи
df['date'] = pd.to_datetime(dict(year=df['YrSold'], month=df['MoSold'], day=1))
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонки')
df[['date', 'YrSold', 'MoSold', 'SalePrice']].head()
```

## 3. Сбор временного ряда

Группирую данные по месяцу и считаю среднюю цену продажи. Если какого-то месяца нет, заполняю его интерполяцией, чтобы ряд был регулярным.

```python
# Считаю среднюю цену и количество продаж по месяцам
month_df = df.groupby('date').agg(mean_price=('SalePrice', 'mean'), sales_count=('SalePrice', 'size')).sort_index()
# Делаю месячную частоту и заполняю пропущенные месяцы
month_df = month_df.asfreq('MS')
month_df['mean_price'] = month_df['mean_price'].interpolate()
month_df['sales_count'] = month_df['sales_count'].fillna(0)
# Основной ряд для прогноза
ts = month_df['mean_price']
print(f'Начало ряда: {ts.index.min().date()}')
print(f'Конец ряда:  {ts.index.max().date()}')
print(f'Количество месяцев: {len(ts)}')
month_df.head()
```

**Короткий вывод:** получился регулярный месячный ряд. Он короткий, поэтому прогнозы нужно воспринимать осторожно: данных всего несколько лет.

## 4. Визуализация ряда

Смотрю, как менялась средняя цена продажи по месяцам. Также рядом показываю количество продаж, потому что месяцы с малым числом сделок могут быть более шумными.

```python
# Строю график средней цены по месяцам
plt.plot(month_df.index, month_df['mean_price'], marker='o', color='steelblue')
plt.title('Средняя цена продажи по месяцам')
plt.xlabel('Дата')
plt.ylabel('Средняя цена')
plt.tight_layout()
plt.show()
```

```python
# Строю график количества продаж по месяцам
plt.bar(month_df.index, month_df['sales_count'], color='salmon', width=20)
plt.title('Количество продаж по месяцам')
plt.xlabel('Дата')
plt.ylabel('Количество продаж')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** ряд заметно шумный, потому что в каждом месяце разное число продаж. Поэтому полезно сравнить несколько подходов, а не полагаться на одну модель.

## 5. Train и test

Последние 12 месяцев оставляю для проверки прогноза. Модели обучаются на более ранней части ряда.

```python
# Последние 12 месяцев оставляю для теста
train = ts.iloc[:-12]
test = ts.iloc[-12:]
print(f'train: {train.index.min().date()} - {train.index.max().date()}, месяцев: {len(train)}')
print(f'test:  {test.index.min().date()} - {test.index.max().date()}, месяцев: {len(test)}')
```

```python
# Показываю train и test на одном графике
plt.plot(train.index, train, label='train', color='steelblue')
plt.plot(test.index, test, label='test', color='coral')
plt.title('Разделение временного ряда')
plt.xlabel('Дата')
plt.ylabel('Средняя цена')
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. STL-декомпозиция

STL разделяет ряд на тренд, сезонную часть и остаток. Это помогает понять, есть ли в данных повторяющийся месячный паттерн.

```python
# Делаю STL-декомпозицию с годовой сезонностью 12 месяцев
stl = STL(ts, period=12)
res_stl = stl.fit()
# Рисую компоненты ряда
res_stl.plot()
plt.tight_layout()
plt.show()
```

**Короткий вывод:** декомпозиция помогает отделить общий уровень цен от сезонных колебаний. На таком коротком ряде сезонность лучше считать ориентиром, а не строгим законом.

## 7. ADF-тест

ADF-тест проверяет стационарность ряда. Если p-value маленький, ряд больше похож на стационарный. Если p-value большой, в ряду может быть тренд или меняющийся уровень.

```python
# Запускаю ADF-тест для исходного ряда
adf = adfuller(ts)
print(f'ADF statistic: {adf[0]:.4f}')
print(f'p-value: {adf[1]:.4f}')
```

**Короткий вывод:** p-value нужно сравнивать с 0.05. Если оно больше 0.05, ряд нельзя уверенно считать стационарным.

## 8. Наивный прогноз

Сначала делаю простой baseline: прогноз равен последнему известному значению train. С ним удобно сравнивать более сложные модели.

```python
# Наивный прогноз: все будущие месяцы равны последнему значению train
naive_pred = pd.Series(train.iloc[-1], index=test.index)
# Считаю качество baseline
naive_mae = mean_absolute_error(test, naive_pred)
naive_rmse = np.sqrt(mean_squared_error(test, naive_pred))
print(f'Naive MAE: {naive_mae:.2f}')
print(f'Naive RMSE: {naive_rmse:.2f}')
```

## 9. SARIMAX

SARIMAX - классическая модель временных рядов. Здесь я использую простую настройку с годовой сезонностью, чтобы не усложнять ноутбук перебором параметров.

```python
# Обучаю SARIMAX на train
sarimax = SARIMAX(train, order=(1, 1, 1), seasonal_order=(1, 0, 1, 12), enforce_stationarity=False, enforce_invertibility=False)
sarimax_fit = sarimax.fit(disp=False)
# Делаю прогноз на длину test
sarimax_pred = sarimax_fit.forecast(steps=len(test))
# Считаю качество
sarimax_mae = mean_absolute_error(test, sarimax_pred)
sarimax_rmse = np.sqrt(mean_squared_error(test, sarimax_pred))
print(f'SARIMAX MAE: {sarimax_mae:.2f}')
print(f'SARIMAX RMSE: {sarimax_rmse:.2f}')
```

## 10. Prophet

Prophet - модель для временных рядов с трендом и сезонностью. В этом окружении он может быть нестабильным при запуске через Jupyter, поэтому я оставляю его как опциональный блок и по умолчанию не запускаю.

```python
if has_prophet and run_prophet:
    from prophet import Prophet
    p_train = train.reset_index()
    p_train.columns = ['ds', 'y']
    model = Prophet(yearly_seasonality=True, weekly_seasonality=False, daily_seasonality=False)
    model.fit(p_train)
    future = model.make_future_dataframe(periods=len(test), freq='MS')
    forecast = model.predict(future)
    prophet_pred = forecast.set_index('ds')['yhat'].iloc[-len(test):]
    prophet_pred.index = test.index
    prophet_mae = mean_absolute_error(test, prophet_pred)
    prophet_rmse = np.sqrt(mean_squared_error(test, prophet_pred))
    print(f'Prophet MAE: {prophet_mae:.2f}')
    print(f'Prophet RMSE: {prophet_rmse:.2f}')
else:
    prophet_pred = None
    prophet_mae = np.nan
    prophet_rmse = np.nan
    print('Prophet не запускается автоматически. Основное сравнение идет без него.')
```

## 11. ML-подход с лагами

Для ML-модели превращаю временной ряд в обычную таблицу: добавляю предыдущие значения, скользящее среднее и номер месяца.

```python
# Создаю таблицу с лаговыми признаками
ml = pd.DataFrame({'y': ts})
ml['lag_1'] = ml['y'].shift(1)
ml['lag_2'] = ml['y'].shift(2)
ml['lag_3'] = ml['y'].shift(3)
ml['lag_12'] = ml['y'].shift(12)
ml['roll_3'] = ml['y'].shift(1).rolling(3).mean()
ml['month'] = ml.index.month
# Убираю строки, где лаги еще не появились
ml = ml.dropna()
ml.head()
```

```python
# Разделяю лаговую таблицу на train и test по тем же датам
ml_train = ml.loc[ml.index < test.index.min()]
ml_test = ml.loc[ml.index >= test.index.min()]
x_train = ml_train.drop(columns=['y'])
y_train = ml_train['y']
x_test = ml_test.drop(columns=['y'])
y_test = ml_test['y']
print(f'x_train: {x_train.shape}')
print(f'x_test:  {x_test.shape}')
```

```python
# Обучаю RandomForest на лаговых признаках
rf = RandomForestRegressor(n_estimators=200, max_depth=4, random_state=42)
rf.fit(x_train, y_train)
# Делаю прогноз и считаю качество
rf_pred = pd.Series(rf.predict(x_test), index=y_test.index)
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
print(f'RandomForest lags MAE: {rf_mae:.2f}')
print(f'RandomForest lags RMSE: {rf_rmse:.2f}')
```

## 12. Сравнение моделей

Собираю все результаты в одну таблицу. Чем меньше MAE и RMSE, тем лучше прогноз.

```python
# Собираю результаты моделей
score = pd.DataFrame({
    'model': ['Naive', 'SARIMAX', 'Prophet', 'RandomForest lags'],
    'mae': [naive_mae, sarimax_mae, prophet_mae, rf_mae],
    'rmse': [naive_rmse, sarimax_rmse, prophet_rmse, rf_rmse]
})
# Убираю Prophet, если он не запускался
score = score.dropna().sort_values('rmse')
score
```

```python
# Строю график RMSE
sns.barplot(data=score, x='rmse', y='model', color='steelblue')
plt.title('Сравнение моделей по RMSE')
plt.xlabel('RMSE')
plt.ylabel('Модель')
plt.tight_layout()
plt.show()
```

## 13. График прогнозов

Показываю реальные значения test и прогнозы моделей на одном графике.

```python
# Рисую реальные значения и прогнозы
plt.plot(train.index, train, label='train', color='gray')
plt.plot(test.index, test, label='test', color='black')
plt.plot(test.index, naive_pred, label='Naive', linestyle='--')
plt.plot(test.index, sarimax_pred, label='SARIMAX', linestyle='--')
plt.plot(rf_pred.index, rf_pred, label='RandomForest lags', linestyle='--')
if prophet_pred is not None:
    plt.plot(prophet_pred.index, prophet_pred, label='Prophet', linestyle='--')
plt.title('Сравнение прогнозов')
plt.xlabel('Дата')
plt.ylabel('Средняя цена')
plt.legend()
plt.tight_layout()
plt.show()
```

## Итоги

В этом ноутбуке я построила временной ряд средней цены продажи домов по месяцам.

- Из исходных данных можно собрать ряд по дате продажи.
- Ряд короткий и шумный, поэтому прогнозы нужно оценивать осторожно.
- STL помогает увидеть тренд, сезонность и остаток.
- ADF-тест дает быструю проверку стационарности.
- Наивный прогноз нужен как baseline.
- SARIMAX использует структуру временного ряда, а Prophet оставлен как опциональный подход.
- ML-подход с лагами превращает задачу в обычную регрессию.

Главный вывод: для временных рядов важно сравнивать модель не только с другими сложными моделями, но и с простым baseline. Если сложная модель не обгоняет baseline, ее нет смысла усложнять.
