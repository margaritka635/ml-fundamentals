# Кросс-валидация и подбор гиперпараметров

**Датасет:** House Prices

В этом ноутбуке я показываю, как оценивать модель через кросс-валидацию и подбирать гиперпараметры. Задача - предсказать цену дома, поэтому это регрессия.

## План работы

1. Загрузить данные и подготовить признаки.
2. Создать целевую переменную в логарифме цены.
3. Оценить базовую модель через KFold.
4. Подобрать параметры через GridSearchCV.
5. Подобрать параметры через RandomizedSearchCV.
6. Добавить Optuna как опциональный блок.
7. Сравнить результаты и сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, регрессии, кросс-валидации и подбора параметров.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, train_test_split, cross_val_score, GridSearchCV, RandomizedSearchCV
from sklearn.metrics import mean_squared_error
try:
    import optuna
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    has_optuna = True
except Exception:
    has_optuna = False
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
print('Optuna доступна:', has_optuna)
```

## 2. Загрузка данных

Загружаю таблицу с домами. Цену буду предсказывать в логарифме, потому что исходная цена сильно скошена вправо.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Проверяю размер таблицы
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонка')
df.head()
```

## 3. Подготовка цели и признаков

Отделяю цену от признаков. Id убираю, потому что это просто номер объекта.

```python
# y - логарифм цены дома
y = np.log1p(df['SalePrice'])
# x - признаки для модели
x = df.drop(columns=['Id', 'SalePrice'])
print(f'x: {x.shape}')
print(f'y: {y.shape}')
```

## 4. Заполнение пропусков

Для числовых признаков использую медиану, для текстовых - значение None. Это простой вариант подготовки данных для сравнения методов подбора параметров.

```python
# Разделяю признаки на числовые и текстовые
num_cols = x.select_dtypes(include='number').columns
cat_cols = x.select_dtypes(include='object').columns
print(f'Числовых колонок: {len(num_cols)}')
print(f'Текстовых колонок: {len(cat_cols)}')
```

```python
# Числовые пропуски заполняю медианой
x[num_cols] = x[num_cols].fillna(x[num_cols].median())
```

```python
# Текстовые пропуски заполняю словом None
x[cat_cols] = x[cat_cols].fillna('None')
```

```python
# Кодирую текстовые признаки
x = pd.get_dummies(x, drop_first=True)
print(f'Размер x после кодирования: {x.shape}')
print(f'Осталось пропусков: {x.isna().sum().sum()}')
```

## 5. Train и validation

Отдельную validation часть оставляю для финальной проверки. Кросс-валидацию и подбор параметров делаю только на train.

```python
# 80 процентов данных идут на train, 20 процентов - на validation
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42)
print(f'x_train: {x_train.shape}')
print(f'x_val:   {x_val.shape}')
```

## 6. Базовая кросс-валидация

KFold делит train на несколько частей. Модель несколько раз обучается на одних частях и проверяется на оставшейся. Так оценка качества получается стабильнее, чем по одному разбиению.

```python
# Создаю KFold с перемешиванием данных
cv = KFold(n_splits=5, shuffle=True, random_state=42)
# Базовая модель без подбора параметров
base = RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1)
# Считаю RMSE на кросс-валидации. В sklearn ошибка отрицательная, поэтому ставлю минус
scores = -cross_val_score(base, x_train, y_train, cv=cv, scoring='neg_root_mean_squared_error')
print(f'RMSE по фолдам: {scores.round(4)}')
print(f'Средний RMSE: {scores.mean():.4f}')
print(f'Стандартное отклонение: {scores.std():.4f}')
```

**Короткий вывод:** кросс-валидация показывает не одно случайное качество, а несколько оценок. Если разброс по фолдам небольшой, модель ведет себя стабильнее.

## 7. GridSearchCV

GridSearchCV перебирает все комбинации параметров из заданной сетки. Это понятно, но может быть медленно, если сетка большая.

```python
# Задаю небольшую сетку параметров, чтобы ноутбук быстро запускался
grid_params = {
    'n_estimators': [100, 200],
    'max_depth': [6, 10, None],
    'min_samples_leaf': [1, 3]
}
# Создаю GridSearchCV
grid = GridSearchCV(RandomForestRegressor(random_state=42, n_jobs=-1), grid_params, cv=cv, scoring='neg_root_mean_squared_error', n_jobs=-1)
# Запускаю подбор параметров
grid.fit(x_train, y_train)
print('Лучшие параметры:', grid.best_params_)
print(f'Лучший CV RMSE: {-grid.best_score_:.4f}')
```

## 8. RandomizedSearchCV

RandomizedSearchCV не проверяет все варианты, а случайно выбирает несколько комбинаций. Это удобно, когда параметров много.

```python
# Задаю распределения параметров для случайного поиска
random_params = {
    'n_estimators': [100, 150, 200, 250],
    'max_depth': [4, 6, 8, 10, None],
    'min_samples_leaf': [1, 2, 3, 5],
    'max_features': ['sqrt', 0.5, 0.8]
}
# Создаю RandomizedSearchCV
random_search = RandomizedSearchCV(RandomForestRegressor(random_state=42, n_jobs=-1), random_params, n_iter=10, cv=cv, scoring='neg_root_mean_squared_error', random_state=42, n_jobs=-1)
# Запускаю подбор параметров
random_search.fit(x_train, y_train)
print('Лучшие параметры:', random_search.best_params_)
print(f'Лучший CV RMSE: {-random_search.best_score_:.4f}')
```

## 9. Optuna

Optuna подбирает параметры умнее случайного поиска: она пробует варианты и постепенно фокусируется на более перспективных. Если Optuna не установлена, этот блок просто пропускается.

```python
# Запускаю Optuna только если библиотека доступна
if has_optuna:
    def objective(trial):
        n_estimators = trial.suggest_int('n_estimators', 80, 220)
        max_depth = trial.suggest_int('max_depth', 4, 14)
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 5)
        max_features = trial.suggest_float('max_features', 0.4, 1.0)
        model = RandomForestRegressor(n_estimators=n_estimators, max_depth=max_depth, min_samples_leaf=min_samples_leaf, max_features=max_features, random_state=42, n_jobs=-1)
        score = -cross_val_score(model, x_train, y_train, cv=cv, scoring='neg_root_mean_squared_error').mean()
        return score
    study = optuna.create_study(direction='minimize')
    study.optimize(objective, n_trials=10)
    optuna_rmse = study.best_value
    optuna_params = study.best_params
    print('Лучшие параметры Optuna:', optuna_params)
    print(f'Лучший CV RMSE Optuna: {optuna_rmse:.4f}')
else:
    optuna_rmse = np.nan
    optuna_params = None
    print('Optuna не установлена, поэтому этот блок пропущен.')
```

## 10. Сравнение способов подбора

Собираю результаты в одну таблицу. Так проще увидеть, какой способ дал лучший RMSE на кросс-валидации.

```python
# Собираю результаты в таблицу
res = pd.DataFrame({
    'method': ['Baseline CV', 'GridSearchCV', 'RandomizedSearchCV', 'Optuna'],
    'cv_rmse': [scores.mean(), -grid.best_score_, -random_search.best_score_, optuna_rmse]
})
# Убираю Optuna из таблицы, если она не запускалась
res = res.dropna().sort_values('cv_rmse')
res
```

```python
# Строю график сравнения RMSE
sns.barplot(data=res, x='cv_rmse', y='method', color='steelblue')
plt.title('Сравнение способов подбора параметров')
plt.xlabel('CV RMSE')
plt.ylabel('Метод')
plt.tight_layout()
plt.show()
```

## 11. Финальная проверка на validation

Беру лучшую модель из RandomizedSearchCV или GridSearchCV и проверяю ее на отложенной validation части.

```python
# Выбираю лучшую модель между GridSearchCV и RandomizedSearchCV
if -random_search.best_score_ <= -grid.best_score_:
    best_model = random_search.best_estimator_
    best_name = 'RandomizedSearchCV'
else:
    best_model = grid.best_estimator_
    best_name = 'GridSearchCV'
# Делаю прогноз на validation
pred = best_model.predict(x_val)
val_rmse = np.sqrt(mean_squared_error(y_val, pred))
print(f'Лучшая модель по поиску: {best_name}')
print(f'Validation RMSE: {val_rmse:.4f}')
```

```python
# Перевожу прогнозы обратно из логарифма в обычные цены для графика
pred_price = np.expm1(pred)
true_price = np.expm1(y_val)
# Строю график реальных цен и прогнозов
plt.scatter(true_price, pred_price, alpha=0.6, color='steelblue')
plt.plot([true_price.min(), true_price.max()], [true_price.min(), true_price.max()], color='red')
plt.title('Реальная цена и прогноз после подбора параметров')
plt.xlabel('Реальная цена')
plt.ylabel('Прогноз')
plt.tight_layout()
plt.show()
```

## Итоги

В этом ноутбуке я показала, как использовать кросс-валидацию и подбор гиперпараметров.

- KFold дает более стабильную оценку качества, чем одно разбиение train validation.
- GridSearchCV перебирает все варианты из сетки, но может быть медленным.
- RandomizedSearchCV проверяет случайные варианты и часто быстрее находит хорошую комбинацию.
- Optuna может искать параметры умнее, но в текущем окружении она может быть не установлена.
- Финальную оценку лучше делать на отдельной validation части, которую не использовали при подборе.

Главный вывод: подбор параметров нужен не ради красивой таблицы, а чтобы честно сравнить варианты модели и не выбирать параметры на глаз.
