# Кластеризация

**Датасет:** House Prices

В этом ноутбуке я группирую дома на похожие сегменты без заранее заданного ответа. Это задача кластеризации: модель не знает правильные классы, а ищет структуру в данных сама.

## План работы

1. Загрузить данные и выбрать признаки для кластеризации.
2. Заполнить пропуски и масштабировать признаки.
3. Подобрать число кластеров для K-Means.
4. Обучить K-Means и описать кластеры.
5. Проверить DBSCAN.
6. Проверить иерархическую кластеризацию.
7. Использовать PCA для визуализации.
8. Сделать выводы.

## 1. Импорт библиотек

Подключаю библиотеки для таблиц, графиков, кластеризации, PCA, масштабирования и оценки качества кластеров.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
%matplotlib inline
pd.set_option('display.max_columns', 120)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)
```

## 2. Загрузка данных

Загружаю данные о домах. Для кластеризации выберу несколько понятных числовых признаков, которые описывают размер, качество, возраст и цену дома.

```python
# Читаю данные из папки data
df = pd.read_csv('../data/train.csv')
# Создаю несколько понятных признаков
df['TotalSF'] = df['TotalBsmtSF'] + df['1stFlrSF'] + df['2ndFlrSF']
df['HouseAge'] = df['YrSold'] - df['YearBuilt']
df['RemodAge'] = df['YrSold'] - df['YearRemodAdd']
print(f'Размер таблицы: {df.shape[0]} строк, {df.shape[1]} колонки')
df.head()
```

## 3. Выбор признаков

Для кластеризации беру только числовые признаки, которые легко интерпретировать. Цена здесь используется не как ответ, а как один из признаков сегментации.

```python
# Список признаков для кластеризации
cols = ['OverallQual', 'OverallCond', 'TotalSF', 'GrLivArea', 'TotalBsmtSF', 'GarageCars', 'GarageArea', 'HouseAge', 'RemodAge', 'SalePrice']
# Создаю рабочую таблицу
x = df[cols].copy()
# Проверяю пропуски
x.isna().sum()
```

```python
# Заполняю возможные пропуски медианой
x = x.fillna(x.median())
# Проверяю основные статистики
x.describe().round(2)
```

**Короткий вывод:** признаки находятся в разных масштабах: цена измеряется сотнями тысяч, а качество - шкалой от 1 до 10. Поэтому перед кластеризацией нужно масштабирование.

## 4. Масштабирование признаков

K-Means, DBSCAN и иерархическая кластеризация зависят от расстояний между объектами. Если не масштабировать признаки, цена и площадь будут доминировать над остальными колонками.

```python
# Создаю scaler
scaler = StandardScaler()
# Масштабирую признаки
x_sc = scaler.fit_transform(x)
print(f'Размер данных после scaling: {x_sc.shape}')
```

## 5. Подбор числа кластеров для K-Means

Проверяю разные значения k. Метод локтя смотрит на inertia, а silhouette показывает, насколько объекты внутри кластеров похожи друг на друга и отличаются от соседних кластеров.

```python
# Проверяю k от 2 до 8
rows = []
for k in range(2, 9):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(x_sc)
    rows.append({'k': k, 'inertia': model.inertia_, 'silhouette': silhouette_score(x_sc, labels)})
k_df = pd.DataFrame(rows)
k_df
```

```python
# Строю графики inertia и silhouette
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].plot(k_df['k'], k_df['inertia'], marker='o')
ax[0].set_title('Метод локтя')
ax[0].set_xlabel('k')
ax[0].set_ylabel('Inertia')
ax[1].plot(k_df['k'], k_df['silhouette'], marker='o', color='coral')
ax[1].set_title('Silhouette score')
ax[1].set_xlabel('k')
ax[1].set_ylabel('Silhouette')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** число кластеров выбирают не только по максимальной метрике, но и по интерпретируемости. Для сегментации домов удобно взять несколько понятных групп.

## 6. K-Means

Обучаю K-Means. В качестве рабочего варианта беру 4 кластера: это обычно дает понятное разделение на более простые, средние, дорогие и крупные дома.

```python
# Обучаю K-Means с 4 кластерами
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['cluster_kmeans'] = kmeans.fit_predict(x_sc)
# Считаю silhouette для выбранного k
km_sil = silhouette_score(x_sc, df['cluster_kmeans'])
print(f'Silhouette K-Means: {km_sil:.3f}')
print(df['cluster_kmeans'].value_counts().sort_index())
```

## 7. Интерпретация кластеров K-Means

Смотрю средние значения признаков внутри каждого кластера. Так можно понять, чем группы отличаются друг от друга.

```python
# Описываю кластеры средними значениями признаков
profile = df.groupby('cluster_kmeans')[cols].mean().round(1)
profile
```

```python
# Строю график средней цены по кластерам
sns.barplot(data=df, x='cluster_kmeans', y='SalePrice', color='steelblue')
plt.title('Средняя цена по кластерам K-Means')
plt.xlabel('Кластер')
plt.ylabel('Средняя цена')
plt.tight_layout()
plt.show()
```

**Короткий вывод:** кластеры можно интерпретировать через среднюю цену, площадь, качество и возраст. Это превращает технические номера кластеров в понятные сегменты домов.

## 8. PCA для визуализации

PCA сжимает много признаков до двух координат. Это нужно только для визуализации: сама кластеризация обучалась на всех выбранных признаках.

```python
# Сжимаю данные до двух компонент
pca = PCA(n_components=2, random_state=42)
x_pca = pca.fit_transform(x_sc)
# Сохраняю компоненты в таблицу
pca_df = pd.DataFrame({'pc1': x_pca[:, 0], 'pc2': x_pca[:, 1], 'cluster': df['cluster_kmeans']})
print(f'Доля объясненной дисперсии: {pca.explained_variance_ratio_.round(3)}')
```

```python
# Строю кластеры на плоскости PCA
sns.scatterplot(data=pca_df, x='pc1', y='pc2', hue='cluster', palette='tab10', alpha=0.8)
plt.title('K-Means кластеры в PCA-пространстве')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend(title='Кластер')
plt.tight_layout()
plt.show()
```

## 9. DBSCAN

DBSCAN ищет плотные области и может выделять выбросы. Для него не нужно заранее задавать число кластеров, но нужно подобрать eps.

```python
# Обучаю DBSCAN
dbscan = DBSCAN(eps=1.8, min_samples=10)
df['cluster_dbscan'] = dbscan.fit_predict(x_sc)
# Смотрю размеры кластеров. Класс -1 означает выбросы
df['cluster_dbscan'].value_counts().sort_index()
```

```python
# Считаю silhouette только если получилось больше одного нормального кластера
db_labels = df['cluster_dbscan']
normal_mask = db_labels != -1
normal_clusters = db_labels[normal_mask].nunique()
if normal_clusters > 1:
    db_sil = silhouette_score(x_sc[normal_mask], db_labels[normal_mask])
    print(f'Silhouette DBSCAN без выбросов: {db_sil:.3f}')
else:
    db_sil = np.nan
    print('DBSCAN не нашел несколько нормальных кластеров при этих параметрах.')
```

**Короткий вывод:** DBSCAN полезен для поиска выбросов. Но на табличных данных с разной плотностью он не всегда дает удобные бизнес-сегменты.

## 10. Иерархическая кластеризация

Иерархическая кластеризация тоже работает через расстояния. Здесь задаю 4 кластера, чтобы сравнить ее с K-Means.

```python
# Обучаю иерархическую кластеризацию
agg = AgglomerativeClustering(n_clusters=4)
df['cluster_agg'] = agg.fit_predict(x_sc)
# Считаю качество
agg_sil = silhouette_score(x_sc, df['cluster_agg'])
print(f'Silhouette Agglomerative: {agg_sil:.3f}')
print(df['cluster_agg'].value_counts().sort_index())
```

## 11. Сравнение методов

Собираю silhouette для методов, где его корректно считать. Чем выше silhouette, тем более разделенными выглядят кластеры.

```python
# Собираю результаты
res = pd.DataFrame({
    'method': ['K-Means', 'DBSCAN', 'Agglomerative'],
    'silhouette': [km_sil, db_sil, agg_sil]
})
res = res.dropna().sort_values('silhouette', ascending=False)
res
```

```python
# Строю график сравнения
sns.barplot(data=res, x='silhouette', y='method', color='steelblue')
plt.title('Сравнение методов по silhouette')
plt.xlabel('Silhouette')
plt.ylabel('Метод')
plt.tight_layout()
plt.show()
```

## Итоги

В этом ноутбуке я сделала кластеризацию домов по ключевым признакам.

- Для кластеризации важно выбирать признаки, которые реально описывают объект.
- Масштабирование обязательно, потому что методы используют расстояния.
- K-Means требует заранее выбрать число кластеров.
- Silhouette и метод локтя помогают подобрать k, но итоговый выбор должен быть еще и понятным.
- PCA помогает визуализировать кластеры на плоскости.
- DBSCAN может находить выбросы, но не всегда дает удобные сегменты.
- Иерархическая кластеризация дает еще один способ группировки объектов.

Главный вывод: кластеризация полезна для сегментации, но номера кластеров сами по себе ничего не значат. Их нужно описывать через средние значения признаков и переводить в понятные группы.
